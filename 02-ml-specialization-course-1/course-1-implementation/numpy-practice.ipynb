### **Dataset:**
```python
import numpy as np

X = np.array([

    [1200, 3],
    [1500, 3],
    [1800, 4],
    [2000, 5],
```

Assume:
- First column: house size in square feet.
- Second column: number of bedrooms.
- Target: house price in thousands of dollars.

## **Solution:**

````python
"""
A1. Dataset inspection
1. Print the shape of X and y.
2. Find the number of training examples and features.
3. Extract the first training example.
4. Extract all house sizes.
5. Calculate the mean house size.
6. Calculate the maximum and minimum house prices.
"""

# given data
import numpy as np

X = np.array([
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4],
    [2000, 5],
    [2200, 4],
    [2500, 5]
])
y = np.array([300, 350, 420, 500, 550, 530, 650])

print("Shape of X is:", X.shape)
print("Shape of y is:", y.shape)

m, n = X.shape
print("Number of training example is:", m)
print("Number of features are:", n)

print("First training example is:", X[0])

all_house_sizes = X[:, 0]
print("All house sizes are:", all_house_sizes)

mean = np.mean(X[:, 0])
print("Mean of all house sizes is:",round(mean, 2))

print("Maximum house price is:", np.max(y))
print("Minimum house price is:", np.min(y))   
````

````python
"""
A2. Mathematical operations
1. Calculate the dot product between:

`w = np.array([0.2, 10])`

and the first training example.
2. Add a bias of 50.
3. Calculate predictions for every example using NumPy.
4. Calculate the residuals (prediction minus actual value).
"""

w = np.array([0.2, 10])

# Dot product with the first training example 
wx = np.dot(X[0], w)
print("The dot product between w and the first training example is:", wx)

# adding bias
b = 50
prediction = wx + b
print("Prediction with bias:", prediction)

# We need to find the prediction for every feature X along with weight w and bias b
wx_i = np.dot(X, w)
y_hat = wx_i + b
print("The prediction for all house prices are:",y_hat)

# now find the error
print("The error y_hat - y is:", y_hat - y)
````

````python
"""
A3. Data visualization
Create:
- A scatter plot showing house size against price.
- A scatter plot showing bedrooms against price.
Explain which feature appears to have a stronger relationship with house price and why visual correlation alone does not establish causation.
"""
````

````python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 2, figsize=(12,5))

# House size vs price
axes[0].scatter(X[:, 0], y, color='green', s=70)
axes[0].set_title("House Size vs. Price")
axes[0].set_xlabel("House Size (sq ft)")
axes[0].set_ylabel("House Price ($1000s)")
axes[0].grid(True, linestyle='--', alpha=0.6)

# House rooms vs price
axes[1].scatter(X[:, 1], y, color='red', s=70, marker='x')
axes[1].set_title("House Rooms vs. Price")
axes[1].set_xlabel("Rooms")
axes[1].set_ylabel("House Price ($1000s)")
axes[1].grid(True, linestyle='--', alpha=0.6)
plt.show()
````
