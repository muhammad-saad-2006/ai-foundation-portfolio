# `Unsupervised Learning:`

## 1. What is Unsupervised Learning: 

> Unsupervised Learning is the machine learning technique in which we provide unlabelled data and the algorithm has to identify hidden patterns, structure and groups in data.

## 2. Main difference between Supervised and Unsupervised Learning:

| Supervised Learning | Unsupervised Learning |
| :--- | :--- |
| Uses labeled data, meaning every input comes with a known, correct output tag. | Uses unlabeled data, meaning the model only gets raw input features without any answers. |
| Learns a mapping function to predict outcomes for new, unseen data.| Analyzes data to discover hidden patterns, structures, or groupings on its own. |
| Acts like learning with a teacher; the model checks its predictions against the true labels and corrects errors. | Operates without supervision or direct feedback; there is no "right answer" to check against. |
| Used for Classification (e.g., spam vs. not spam) and Regression (e.g., predicting house prices). | Used for Clustering (e.g., customer segmentation) and Association (e.g., market basket analysis). |
| Uses Linear Regression, Logistic Regression, Decision Trees, Support Vector Machines (SVM), and Random Forest. | Uses K-Means Clustering, Hierarchical Clustering, Principal Component Analysis (PCA), and Apriori |

## 3. Types of Unsupervised Learning:

Primarily there are three types of unsupervised learning which are as follows:
- Clustering
- Dimensionality reduction
- Anomaly Detection

### 3.1 Clustering:

> Clustering is an unsupervised machine learning technique that automatically groups similar data points together so that items in the same group (a cluster) are more alike than items in other groups.

**For Example:**
- Customer Segmentation 
- Recommendation Engines
- Image Segmentation & Compression
- Document & News Grouping
