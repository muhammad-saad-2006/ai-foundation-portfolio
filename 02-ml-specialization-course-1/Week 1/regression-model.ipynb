# Linear Regression Model:

## 1. What is linear Regression:

>Linear regression is a supervised machine learning algorithm that predicts a continuous numerical value by fitting a straight line to the relationship between input features and a target outcome

**Formula of Linear Regression:**

$$y = mx + b$$

where:
* **y** $\Rightarrow$ dependent variable, i.e., the variable that we need to predict.
* **m** $\Rightarrow$ slope/weight. It tells us how $y$ changes for every one-unit increase in $x$.
* **x** $\Rightarrow$ independent variable, i.e., the variable on the basis of which we make predictions.
* **b** $\Rightarrow$ intercept/bias. The starting value of $y$ when $x$ is $0$.


**Examples:**
- Real Estate: Predicting the selling price of a house based on its size, number of bedrooms, and location.
- Sales Forecasting: Estimating future product revenue based on advertising spend.
- Academics: Predicting a student's final exam score based on the number of hours studied.
- Healthcare: Estimating blood pressure or cholesterol levels based on age and weight.

## 2. Cost Function Formula:

### 2.1. What is a Cost Function?

A cost function measures how well a machine learning model's predictions match the actual answers in the training dataset.

For linear regression, the model predicts an output using a straight-line equation. The cost function measures the overall prediction error across all training examples.

- **Small cost:** The model's predictions are close to the actual values.
- **Large cost:** The model's predictions are far from the actual values.

**Goal:** Find the values of model parameters that minimize the cost.

### 2.2. Linear Regression Model

For one input feature, the linear regression model is:

$$
f_{w,b}(x) = wx + b
$$

Where:

| Symbol | Meaning |
|---|---|
| $x$ | Input feature |
| $f_{w,b}(x)$ | Model's prediction |
| $w$ | Weight or slope |
| $b$ | Bias or y-intercept |

For the $i^{th}$ training example:

$$
\hat{y}^{(i)} = f_{w,b}(x^{(i)}) = wx^{(i)} + b
$$

- $\hat{y}^{(i)}$ represents the predicted output.
- $y^{(i)}$ represents the actual output.

The values of $w$ and $b$ determine the position and slope of the prediction line.

### 2.3. Calculate the Prediction Error

For one training example, the prediction error is:

$$
e^{(i)} = \hat{y}^{(i)} - y^{(i)}
$$

In simple words:

$$
\text{Error} = \text{Predicted Value} - \text{Actual Value}
$$

**Example 1:**

- Actual value = 300
- Predicted value = 330

$$
Error = 330 - 300 = 30
$$

**Example 2:**

- Actual value = 300
- Predicted value = 270

$$
Error = 270 - 300 = -30
$$

Errors can be positive or negative. If we simply add errors, positive and negative values may cancel each other.

Therefore, we use **Squared Error**:

$$
(e^{(i)})^2 = (\hat{y}^{(i)} - y^{(i)})^2
$$

Squaring makes the error non-negative and gives larger errors a stronger penalty.

### 2.4. Cost Function Formula

For $m$ training examples, the linear regression cost function is:

$$
\boxed{
J(w,b) = \frac{1}{2m}\sum_{i=1}^{m}
\left(f_{w,b}(x^{(i)}) - y^{(i)}\right)^2
}
$$

Since:

$$
f_{w,b}(x^{(i)}) = wx^{(i)} + b
$$

We can also write the formula as:

$$
\boxed{
J(w,b) = \frac{1}{2m}\sum_{i=1}^{m}
\left(wx^{(i)} + b - y^{(i)}\right)^2
}
$$

#### Understanding Each Part of the Formula

| Symbol | Meaning |
|---|---|
| $J(w,b)$ | Total cost |
| $w$ | Weight or slope |
| $b$ | Bias or intercept |
| $m$ | Total number of training examples |
| $\sum$ | Summation: add all the values |
| $x^{(i)}$ | Input of the $i^{th}$ example |
| $y^{(i)}$ | Actual output |
| $f_{w,b}(x^{(i)})$ | Predicted output |
| $(...)^2$ | Squared prediction error |
| $\frac{1}{2m}$ | Average squared error with an additional factor of one-half |

#### Why Do We Divide by m?

We divide by $m$ because we want to calculate the average squared error across all training examples rather than just the total error.

#### Why Do We Divide by 2m Instead of m?

The extra 2 is included mainly for mathematical convenience when calculating derivatives in gradient descent.

When differentiating a squared term, a factor of 2 appears. This cancels the $\frac{1}{2}$ in the cost function.

**Important:** The factor of 2 does not change which values of $w$ and $b$ minimize the cost.

### 2.5. Worked Numerical Example

Suppose we have the following training dataset:

| Example | Input (x) | Actual Output (y) |
|---|---:|---:|
| 1 | 1 | 1 |
| 2 | 2 | 2 |
| 3 | 3 | 3 |

Given model parameters:

$$
w = 0.5,\qquad b = 0
$$

The linear regression model is:

$$
f_{w,b}(x) = 0.5x
$$

#### Step 1: Calculate Predictions

| Example | Calculation | Prediction |
|---|---|---:|
| 1 | $0.5 \times 1$ | 0.5 |
| 2 | $0.5 \times 2$ | 1.0 |
| 3 | $0.5 \times 3$ | 1.5 |

#### Step 2: Calculate Squared Errors

Formula:

$$
Squared\ Error = (Prediction - Actual)^2
$$

| Example | Prediction | Actual | Error | Squared Error |
|---|---:|---:|---:|---:|
| 1 | 0.5 | 1 | -0.5 | 0.25 |
| 2 | 1.0 | 2 | -1.0 | 1.00 |
| 3 | 1.5 | 3 | -1.5 | 2.25 |
| **Total** | | | | **3.50** |

#### Step 3: Apply the Cost Function

We have:

$$
m = 3
$$

Cost function:

$$
J(w,b) = \frac{1}{2m}\sum_{i=1}^{m}
(Prediction - Actual)^2
$$

Substitute the values:

$$
J(0.5,0) = \frac{1}{2(3)}(0.25 + 1 + 2.25)
$$

$$
J(0.5,0) = \frac{3.5}{6}
$$

$$
\boxed{J(0.5,0) = 0.5833}
$$

The cost is approximately **0.5833**.

#### Step 4: Compare with a Better Model

Now choose:

$$
w = 1,\qquad b = 0
$$

The model becomes:

$$
f_{w,b}(x) = x
$$

Predictions:

- For $x=1$, prediction = 1
- For $x=2$, prediction = 2
- For $x=3$, prediction = 3

All predictions match the actual values.

Therefore:

$$
J(1,0) = \frac{1}{2(3)}(0^2+0^2+0^2)
$$

$$
\boxed{J(1,0)=0}
$$

**Conclusion:** The second model has a lower cost and perfectly fits the training examples.

### 2.6. What Are We Trying to Find?

The cost function depends on two parameters:

- $w$: Weight or slope
- $b$: Bias or y-intercept

Different values of $w$ and $b$ produce different prediction lines and different cost values.

Our objective is:

$$
\boxed{\underset{w,b}{\text{minimize}}\ J(w,b)}
$$

In simple words:

1. Choose values for $w$ and $b$.
2. Make predictions using the model.
3. Calculate the prediction errors.
4. Square the errors.
5. Calculate the total cost.
6. Find values of $w$ and $b$ that minimize the cost.

**Gradient Descent** is one method used to adjust $w$ and $b` to reduce the cost.

### Problem Statement

You would like a model which can predict housing prices given the size of the house.  
Let's use the same two data points as the previous lab - a house with 1000 square feet sold for \\$300,000 and a house with 2000 square feet sold for \\$500,000.


| Size (1000 sqft)     | Price (1000s of dollars) |
| -------------------| ------------------------ |
| 1                 | 300                      |
| 2                  | 500                      |

### Solution

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline

print("Imports are successful")
```

```python
x_train = np.array([1.0, 2.0])           #(size in 1000 square feet)
y_train = np.array([300.0, 500.0])           #(price in 1000s of dollars)
```

```python
# writing function to calculate cost
def compute_cost(x, y, w, b):
    m = x.shape[0]
    cost_sum = 0

    for i in range(m):
        f_wb = w*x[i] + b
        cost = (f_wb - y[i])**2
        cost_sum = cost_sum + cost
        
    total_cost = (1/(2*m)) * cost_sum
    return total_cost
```

```python
# let us use different values of w and b to see where the cost is minimum
# we are using 3 different values for 
w1 = 100
b1 = 0

w2 = 150
b2 = 50

w3 = 200
b3 = 100

print(f"The cost when w is {w1} and b is {b1}:", compute_cost(x_train, y_train, w1, b1))
print(f"The cost when w is {w2} and b is {b2}:", compute_cost(x_train, y_train, w2, b2))
print(f"The cost when w is {w3} and b is {b3}:", compute_cost(x_train, y_train, w3, b3))

# the result shows by increasing weight and bias we are reaching to minimum cost
```

```python
w_vals = np.linspace(0, 400, 100)
b_vals = np.linspace(-100, 300, 100)
W, B = np.meshgrid(w_vals, b_vals)

# 2. Compute J(w, b) for every pair on the grid using your function
J = np.zeros_like(W)
for i in range(W.shape[0]):
    for j in range(W.shape[1]):
        J[i, j] = compute_cost(x_train, y_train, W[i, j], B[i, j])

# 3. Create side-by-side plots (2D Contour + 3D Surface)
fig = plt.figure(figsize=(13, 5))

# --- Left Plot: 2D Contour Plot ---
ax1 = fig.add_subplot(1, 2, 1)
levels = np.logspace(1, 5, 15)
contours = ax1.contour(W, B, J, levels=levels, cmap='viridis')
ax1.clabel(contours, inline=True, fontsize=8, fmt='%1.0f')
ax1.scatter(200, 100, color='red', marker='x', s=100, zorder=5, label='Minimum (w=200, b=100)')

ax1.set_title("Contour Plot of Cost J(w, b)")
ax1.set_xlabel("Weight (w)")
ax1.set_ylabel("Bias (b)")
ax1.legend()
ax1.grid(True, linestyle='--', alpha=0.5)

# --- Right Plot: 3D "Soup Bowl" Surface Plot ---
ax2 = fig.add_subplot(1, 2, 2, projection='3d')
surf = ax2.plot_surface(W, B, J, cmap='viridis', alpha=0.85, edgecolor='none')

ax2.set_title("3D Surface Plot of Cost J(w, b)")
ax2.set_xlabel("Weight (w)")
ax2.set_ylabel("Bias (b)")
ax2.set_zlabel("Cost J(w, b)")

plt.tight_layout()
plt.show()
```

## 3. Gradient Descent:

### 3.1. What is Gradient Descent:

> Gradient Descent is a optimization technique that is used to minimize the cost the function.

We have recently seen that we can find minimum point if we adjust hour **w** and **b**. But the question is how we are going to adjust them. One way is to manually find the values of **w** and **b** by substituting different values unless we got the minimum cost.

This method stands out when data is small but this become in efficient when dealing with large data it will become a serious headache. So what to do know?

Here gradient descent comes into action it is the technique that will help us to find the local minima of given dataset. Mathematically it is;
$$\min_{w,b} J(w, b)$$

### 3.2. How Does Gradient Descent Work?

Gradient Descent follows a repeated process:

1. Start with initial values of $w$ and $b$.
2. Calculate the current cost.
3. Determine the direction that decreases the cost most quickly.
4. Take a small step in that direction.
5. Update the parameters.
6. Repeat until the cost reaches or approaches a minimum.

**Main idea:**

Keep changing the parameters slightly to reduce the cost function.


Think of it as walking down a hill.

Your goal is to reach the lowest point of the hill.

At every position, you look around and decide which direction takes you downhill most quickly.

That is the basic intuition behind Gradient Descent.

### 3.3. What Is a Local Minimum?

Some cost functions have multiple minimum points. A **local minimum** is a point where the cost is lower than the cost at nearby points. However, it might not be the lowest point across the entire function.

Imagine a landscape containing two valleys:
- Valley 1 is on the left.
- Valley 2 is on the right.

Suppose you start at a position near Valley 1.

Gradient Descent takes you downhill until you reach Valley 1.

Now suppose you restart from a different position near Valley 2.

Gradient Descent takes you downhill until you reach Valley 2.

You can reach different minimum points depending on your starting position.

#### Why Does This Happen?

Gradient Descent follows the direction of steepest descent from its current position.

Once it reaches a local minimum, there may be no nearby downhill direction that takes it out of that valley.

Therefore, it can get stuck in that local minimum instead of reaching another, potentially lower valley.

### 3.4. Local Minimum vs Global Minimum

These two concepts are important.

| Local Minimum | Global Minimum |
|---|---|
| Lowest point in a nearby region | Lowest point across the entire function |
| Other lower points may exist elsewhere | No lower point exists anywhere |
| Can depend on the starting region | Represents the overall lowest cost |

**Important distinction:**

For ordinary Linear Regression with the squared-error cost function, the cost function is convex (bowl-shaped or, in some cases, a flat-bottomed bowl).

This means every local minimum is also a global minimum.

However, more complex models, such as neural networks, can have non-convex cost functions with multiple local minima.


### 3.5. Gradient Descent Formula

Gradient Descent updates model parameters repeatedly to minimize the cost function.

For Linear Regression, we have two parameters: $w$ and $b$.

The update equations are:

$$
\boxed{w := w - \alpha \frac{\partial}{\partial w}J(w,b)}
$$

$$
\boxed{b := b - \alpha \frac{\partial}{\partial b}J(w,b)}
$$

Where:

| Symbol | Meaning |
|---|---|
| $w$ | Weight (slope) |
| $b$ | Bias (intercept) |
| $\alpha$ | Learning rate |
| $J(w,b)$ | Cost function |
| $\frac{\partial J}{\partial w}$ | Derivative of cost with respect to $w$ |
| $\frac{\partial J}{\partial b}$ | Derivative of cost with respect to $b$ |
| $:=$ | Assignment: update the parameter with a new value |

**Main idea:**

New Parameter = Old Parameter - Adjustment

We repeatedly update $w$ and $b$ to reduce the cost function.

#### 3.5.1. Learning Rate ($\alpha$)

The learning rate controls the size of each step taken towards minimizing the cost function.

It is usually a small positive number, for example:

$$
\alpha = 0.01
$$

##### What happens when we change the learning rate?

| Learning Rate | Effect |
|---|---|
| Too Large | Takes huge steps, may overshoot the minimum or diverge |
| Too Small | Takes very small steps and convergence becomes slow |
| Appropriate | Makes suitable progress towards the minimum |

**Remember:** The learning rate controls how big a step we take during Gradient Descent.

#### 3.5.2. Understanding the Derivative Term

The derivative tells us the slope of the cost function at the current parameter value.

For Gradient Descent:

- Positive derivative: The update decreases the parameter.
- Negative derivative: The update increases the parameter.
- Zero derivative: No change occurs in that parameter through its gradient update.

The negative sign in the Gradient Descent formula makes the update move in the direction of decreasing cost.

The derivative and learning rate together determine the parameter adjustment.

#### 3.5.3. When Does Gradient Descent Stop?

Gradient Descent repeatedly updates the parameters until the algorithm converges.

**Convergence** means that the parameter values no longer change significantly with additional iterations.

Mathematically, we look for updates where:

$$
|w_{\text{new}}-w_{\text{old}}| \approx 0
$$

$$
|b_{\text{new}}-b_{\text{old}}| \approx 0
$$

The algorithm has approximately reached a minimum or a point where further progress is very small.

#### 3.5.4. Simultaneous Updates

This is one of the most important implementation concepts in the lecture.

We have two parameters:

- $w$
- $b$

Both must be updated simultaneously using their **old values**.

##### Correct Implementation

First, calculate both new values using the old values of $w$ and $b$.

$$
tempW = w - \alpha \frac{\partial J(w,b)}{\partial w}
$$

$$
tempB = b - \alpha \frac{\partial J(w,b)}{\partial b}
$$

Then update both parameters:

$$
w = tempW
$$

$$
b = tempB
$$

##### **Why do we need simultaneous updates?**

Because both calculations must use the same previous parameter values.

If we update $w$ first and then calculate $b$, the calculation of $b$ may use the new value of $w$ instead of its old value.

This changes the intended Gradient Descent update.

**Important:** Always calculate both updates first, then assign both new values.

#### 3.5.5. Complete Gradient Descent Process

1. Initialize $w$ and $b$ (commonly both zero for Linear Regression).
2. Calculate the derivatives of the cost function.
3. Calculate the new values of $w$ and $b$ using the Gradient Descent equations.
4. Update both parameters simultaneously.
5. Repeat the process until convergence.

### 3.6. Simplifying Gradient Descent

To understand Gradient Descent more easily temporarily considers a cost function with only one parameter:

$$
J(w)
$$

Instead of working with two parameters ($w$ and $b$), we only work with $w$.

The Gradient Descent formula becomes:

$$
\boxed{w := w-\alpha\frac{d}{dw}J(w)}
$$

Our objective is to find the value of $w$ that minimizes the cost function.

#### 3.6.1. Three Cases of Derivatives

#### Case 1: Positive Derivative

When the tangent line slopes upward from left to right:

$$
\frac{dJ(w)}{dw}>0
$$

Example:

$$
\frac{dJ(w)}{dw}=2
$$

Using Gradient Descent:

$$
w_{\text{new}}=w_{\text{old}}-\alpha(2)
$$

Since we subtract a positive number, the value of $w$ decreases.

Therefore, we move towards the **left**.

On the right side of a bowl-shaped cost function, moving left takes us towards a lower cost.

#### Case 2: Negative Derivative

When the tangent line slopes downward from left to right:

$$
\frac{dJ(w)}{dw}<0
$$

Example:

$$
\frac{dJ(w)}{dw}=-2
$$

Using Gradient Descent:

$$
w_{\text{new}}=w_{\text{old}}-\alpha(-2)
$$

Subtracting a negative number is the same as adding a positive number.

Therefore, the value of $w$ increases.

We move towards the **right**.

On the left side of a bowl-shaped cost function, moving right takes us towards a lower cost.

#### Case 3: Zero Derivative

At the bottom of a bowl-shaped cost function, the tangent line is horizontal.

$$
\frac{dJ(w)}{dw}=0
$$

Using Gradient Descent:

$$
w_{\text{new}}=w_{\text{old}}-\alpha(0)
$$

$$
\boxed{w_{\text{new}}=w_{\text{old}}}
$$

The parameter does not change.

This is the minimum point in our simple bowl-shaped example.

### 3.6.2. Summary of the Three Cases

| Derivative | Parameter Update | Direction |
|---|---|---|
| Positive (+) | Subtract a positive value | Move left (decrease $w$) |
| Negative (-) | Subtract a negative value | Move right (increase $w$) |
| Zero (0) | Subtract zero | No change in $w$ |

**Important:** Gradient Descent automatically determines whether to increase or decrease the parameter based on the derivative.

### 3.6.3. Worked Numerical Example

Suppose our cost function is:

$$
J(w)=(w-3)^2
$$

Its minimum is at:

$$
w=3
$$

The derivative is:

$$
\frac{dJ(w)}{dw}=2(w-3)
$$

Choose the learning rate:

$$
\alpha=0.1
$$

#### Situation A: Starting on the Right Side

Initial value:

$$
w=5
$$

Calculate the derivative:

$$
J'(5)=2(5-3)=4
$$

Update the parameter:

$$
w_{\text{new}}=5-(0.1)(4)
$$

$$
\boxed{w_{\text{new}}=4.6}
$$

The derivative is positive, so $w$ decreases and moves left.

Cost before update:

$$
J(5)=(5-3)^2=4
$$

Cost after update:

$$
J(4.6)=(4.6-3)^2=2.56
$$

The cost decreased from 4 to 2.56.

#### Situation B: Starting on the Left Side

Initial value:

$$
w=1
$$

Calculate the derivative:

$$
J'(1)=2(1-3)=-4
$$

Update the parameter:

$$
w_{\text{new}}=1-(0.1)(-4)
$$

$$
\boxed{w_{\text{new}}=1.4}
$$

The derivative is negative, so $w$ increases and moves right.

Cost before update:

$$
J(1)=(1-3)^2=4
$$

Cost after update:

$$
J(1.4)=(1.4-3)^2=2.56
$$

The cost decreased from 4 to 2.56.

#### Situation C: Reaching the Minimum

Suppose:

$$
w=3
$$

Calculate the derivative:

$$
J'(3)=2(3-3)=0
$$

Update:

$$
w_{\text{new}}=3-(0.1)(0)
$$

$$
\boxed{w_{\text{new}}=3}
$$

The parameter stops changing because the derivative is zero.

### 3.6.4. Important Clarification

A negative derivative does not mean the cost function is negative.

It means that the cost is decreasing as $w$ increases.

The derivative and cost are two different things:

- Cost $J(w)$ tells us the value of the cost function.
- Derivative $J'(w)$ tells us its slope at a particular point.

Also, a zero derivative does not always mean a minimum. It can also occur at a maximum or saddle point.
