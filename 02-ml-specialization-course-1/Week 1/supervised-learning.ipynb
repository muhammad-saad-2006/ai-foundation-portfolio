# `Supervised Learning:`

## 1. What is Supervised Learning:
>"Supervised learning is a machine learning technique in which we have labelled data and we use this data as input and than train our machine to learn that data make patterns and eventually become smarter so that it can accurately predict any incoming unseen data." 

## 2. Types of Supervised Learning:

There are primarily two types of supervised learning:
* Regression
* Classification

### 2.1 Regression:

> It is machine learning algorithm in which model learns to predicts numbers.

**For Example:**
- House Price Prediction
- Insurance Claim
- Match Score Prediction and many more

### 2.2 Classification:

> It is a machine learning algorithm in which machine learns to predict discrete categories.

**For Example:**
- Spam Email Classification
- Gender Classification
- Breast Cancer Detection and many more
