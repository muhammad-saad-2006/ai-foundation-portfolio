# **Polynomial Regression and Feature Engineering**

## 1. Introduction

Until now, we have used Linear Regression to fit straight lines to training data.

However, real-world datasets do not always follow a straight-line relationship.

For example, house prices may increase in a nonlinear way as house size increases.

A straight line may not provide a good fit for such data.

To solve this problem, we introduce **Polynomial Regression**.

Polynomial Regression allows Linear Regression to fit nonlinear curves by creating additional features using powers or transformations of existing features.

## 2. Ordinary Linear Regression

Suppose we want to predict house prices using one feature:

- $x$ = House size in square feet.
- $y$ = House price.

The ordinary Linear Regression model is:

$$
\boxed{f_{w,b}(x)=w_1x+b}
$$

This model fits a straight line to the training data.

However, if the relationship between house size and price is curved, a straight-line model may not fit the data well.

Therefore, we need a more flexible model.

## 3. Polynomial Regression

Polynomial Regression creates additional features by raising the original feature to different powers.

Instead of using only $x$, we can also use:

$$
x^2,\quad x^3,\quad x^4,\ldots
$$

These are called polynomial features.

Andrew Ng introduces three main models.

### A. Linear Model

The original model uses only one feature:

$$
\boxed{f(x)=w_1x+b}
$$

This produces a straight line.

### B. Quadratic Model

We create a second feature by squaring the original feature:

$$
x^2
$$

The model becomes:

$$
\boxed{f(x)=w_1x+w_2x^2+b}
$$

This is called a Quadratic Model because the highest power of the input is 2.

It can fit a curved shape called a parabola.

**Important:** If the coefficient of $x^2$ is negative, the parabola opens downward and eventually turns downward.

If the coefficient is positive, the parabola opens upward.

### C. Cubic Model

We create another feature:

$$
x^3
$$

The model becomes:

$$
\boxed{f(x)=w_1x+w_2x^2+w_3x^3+b}
$$

This is called a Cubic Model because the highest power is 3.

A cubic model can represent more complicated curves, including curves that change direction.

The exact shape depends on the learned parameter values.

### Comparison

| Model | Equation | Highest Power |
|-------|----------|---------------|
| Linear | $w_1x+b$ | 1 |
| Quadratic | $w_1x+w_2x^2+b$ | 2 |
| Cubic | $w_1x+w_2x^2+w_3x^3+b$ | 3 |

## 4. Connection Between Polynomial Regression and Multiple Linear Regression

Polynomial Regression is actually an application of Multiple Linear Regression using engineered features.

Recall the Multiple Linear Regression model:

$$
f_{\vec{w},b}(\vec{x})=
w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

In Polynomial Regression, we use transformed versions of the original feature.

For a cubic model, we have:

$$
\vec{x}=[x,x^2,x^3]
$$

The weight vector is:

$$
\vec{w}=[w_1,w_2,w_3]
$$

Therefore, the model becomes:

$$
\boxed{
f_{\vec{w},b}(\vec{x})=
\vec{w}\cdot\vec{x}+b
}
$$

Which is equivalent to:

$$
f(x)=w_1x+w_2x^2+w_3x^3+b
$$

### Numerical Example

Suppose:

$$
x=2
$$

Our engineered feature vector becomes:

$$
\vec{x}=[2,2^2,2^3]
$$

$$
\vec{x}=[2,4,8]
$$

Suppose the weights are:

$$
\vec{w}=[3,2,0.5]
$$

And:

$$
b=5
$$

Using the polynomial regression model:

$$
f(x)=3x+2x^2+0.5x^3+5
$$

Substitute $x=2$:

$$
f(2)=3(2)+2(4)+0.5(8)+5
$$

$$
=6+8+4+5
$$

$$
\boxed{f(2)=23}
$$

This demonstrates how polynomial regression uses multiple features to calculate predictions.

**Important Mathematical Concept:**

Polynomial Regression is nonlinear with respect to the input feature $x$, but it remains linear with respect to its parameters (weights).

This allows us to use Linear Regression techniques, including Gradient Descent, to train the model.

## 5. Feature Engineering

Feature Engineering is the process of creating, transforming or combining existing features to provide more useful information to a machine learning model.

In our example, the original dataset contains only one feature:

$$
x=\text{House Size}
$$

We create additional features:

$$
x^2=\text{House Size Squared}
$$

$$
x^3=\text{House Size Cubed}
$$

Our new feature set becomes:

$$
\boxed{\vec{x}=[x,x^2,x^3]}
$$

We have transformed one original feature into three features.

This process is called Feature Engineering.

### Why Is Feature Engineering Important?

1. It provides additional information to the model.
2. It allows Linear Regression to fit nonlinear relationships.
3. It makes the model more flexible.
4. It can improve predictions when the engineered features represent useful patterns in the data.

**Remember:** Feature Engineering creates new features from existing information. It does not necessarily require collecting new data.

## 6. Using the Square Root as a Feature

Instead of using:

$$
x,\quad x^2,\quad x^3
$$

We can use:

$$
x,\quad \sqrt{x}
$$

Our model becomes:

$$
\boxed{f(x)=w_1x+w_2\sqrt{x}+b}
$$

The square root function has a different shape from a straight line or polynomial curve.

### Understanding the Square Root Function

Consider:

$$
y=\sqrt{x}
$$

| $x$ | $\sqrt{x}$ |
|---:|---:|
| 1 | 1 |
| 4 | 2 |
| 9 | 3 |
| 16 | 4 |
| 25 | 5 |
| 100 | 10 |

As $x$ increases, its square root also increases, but its rate of increase becomes smaller.

The curve becomes less steep as $x$ increases, but it does not turn downward.

The slope of the square root function is:

$$
\frac{d}{dx}\sqrt{x}=
\frac{1}{2\sqrt{x}}
$$

For positive values of $x$, the slope is positive and approaches zero as $x$ increases.

**Important:** The square root transformation requires nonnegative input values when working with real numbers.

## 7. Feature Scaling With Polynomial Features

Feature Scaling becomes increasingly important when we create polynomial features.

Consider a house-size feature ranging from:

$$
1\text{ to }1000
$$

### Original Feature

$$
x=1\text{ to }1000
$$

### Second Feature: Squared

$$
x^2
$$

Its range becomes:

$$
1^2\text{ to }1000^2
$$

$$
\boxed{1\text{ to }1,000,000}
$$

### Third Feature: Cubed

$$
x^3
$$

Its range becomes:

$$
1^3\text{ to }1000^3
$$

$$
\boxed{1\text{ to }1,000,000,000}
$$

### Comparison

| Feature | Minimum | Maximum |
|---------|---------|---------|
| $x$ | 1 | 1,000 |
| $x^2$ | 1 | 1,000,000 |
| $x^3$ | 1 | 1,000,000,000 |

Notice that these features have very different numerical ranges.

When Gradient Descent works with these unscaled features, it may become slow because the cost function can have poorly balanced contours.

### Solution: Apply Feature Scaling

After creating polynomial features, apply Feature Scaling to make their ranges comparable.

The process is:

1. Start with the original feature.
2. Create polynomial features such as $x^2$ and $x^3$.
3. Apply Feature Scaling separately to each engineered feature.
4. Train the model using Gradient Descent.

Possible scaling techniques include:

- Dividing by the maximum.
- Mean Normalization.
- Z-Score Normalization.

The same training-set scaling statistics must be reused for validation, test and future data.

**Important:** Feature Engineering should happen before scaling the newly created features.

## 8. How Do We Choose Which Features to Use?

We have several possible feature choices:

- Only $x$.
- $x$ and $x^2$.
- $x$, $x^2$ and $x^3$.
- $x$ and $\sqrt{x}$.

How do we decide which features are best?

There is no single answer that works for every dataset.

The choice depends on the relationship between the input features and the target output.

To choose suitable features:

1. Create different possible feature sets.
2. Train different models using those features.
3. Evaluate model performance on unseen data.
4. Compare the results.
5. Choose the model that generalizes well.
