# **Multiple Linear Regression:**

## 1. Introduction:

In simple linear regression, we use only one feature (input variable) to predict an output.

For example, predicting house prices using only the size of the house.

The model is:

$$
f_{w,b}(x) = wx + b
$$

However, house prices depend on multiple factors, such as:
- House size
- Number of bedrooms
- Number of floors
- Age of the house

To use multiple features for prediction, we use **Multiple Linear Regression**.

Multiple Linear Regression is an extension of simple linear regression that uses multiple input features to predict an output.


## 2. Notation for Multiple Features:

Consider a dataset containing four features:

| Feature | Description |
|---------|-------------|
| $x_1$ | House size (square feet) |
| $x_2$ | Number of bedrooms |
| $x_3$ | Number of floors |
| $x_4$ | Age of the house (years) |

### 2.1. Number of Features ($n$):

The lowercase $n$ represents the total number of features in a dataset.

For our example:

$$
n = 4
$$

### 2.2. Training Example ($x^{(i)}$):

A training example represents one complete observation in the dataset.

For example, the second training example can be represented as:

$$
x^{(2)} = [1416, 3, 2, 40]
$$

This means:
- House size = 1416 square feet
- Bedrooms = 3
- Floors = 2
- Age = 40 years

**Important:** The superscript $(i)$ represents the training example number, not an exponent.

### 2.3. Individual Feature of a Training Example ($x_j^{(i)}$):

To access a particular feature of a specific training example, we use:

$$
x_j^{(i)}
$$

Where:
- $i$ = Training example number.
- $j$ = Feature number.

For example:

$$
x_3^{(2)} = 2
$$

This means the third feature (number of floors) of the second training example has a value of 2.

### 2.4. Vector:

A vector is simply an ordered list of numbers.

For example:

$$
\vec{x}^{(2)} = [1416, 3, 2, 40]
$$

This vector contains all four features of the second training example.

A vector can be represented as:
- Row vector: Values arranged horizontally.
- Column vector: Values arranged vertically.

An arrow above a variable (e.g., $\vec{x}$) is sometimes used to indicate that it is a vector rather than a single number. The arrow is optional notation.

## 3. Multiple Linear Regression Model:

In simple linear regression, the model is:

$$
f_{w,b}(x) = wx + b
$$

For multiple features, each feature has its own weight.

For four features, the model becomes:

$$
f_{\vec{w},b}(\vec{x}) =
w_1x_1 + w_2x_2 + w_3x_3 + w_4x_4 + b
$$

Where:

- $x_1, x_2, \ldots, x_4$ = Input features.
- $w_1, w_2, \ldots, w_4$ = Weights corresponding to each feature.
- $b$ = Bias (intercept).
- $f_{\vec{w},b}(\vec{x})$ = Predicted output.

Each feature has a separate weight because different features can have different impacts on the prediction.

### 3.1. Example: House Price Prediction:

Consider the following model:

$$
f_{\vec{w},b}(\vec{x}) =
0.1x_1 + 4x_2 + 10x_3 - 2x_4 + 80
$$

Assume the predicted house price is measured in thousands of dollars.

| Parameter | Value | Interpretation |
|-----------|-------|----------------|
| $w_1$ | 0.1 | Each additional square foot increases the predicted price by $100. |
| $w_2$ | 4 | Each additional bedroom increases the predicted price by $4,000. |
| $w_3$ | 10 | Each additional floor increases the predicted price by $10,000. |
| $w_4$ | -2 | Each additional year of age decreases the predicted price by $2,000. |
| $b$ | 80 | Base prediction is $80,000. |

**Understanding Positive and Negative Weights:**

- A positive weight indicates that increasing the corresponding feature increases the prediction, assuming other features remain constant.
- A negative weight indicates that increasing the corresponding feature decreases the prediction, assuming other features remain constant.

For example, $w_4 = -2$ means that each additional year of house age decreases the predicted price by $2,000.

### Numerical Example:

Suppose a house has the following features:

- Size = 1500 square feet
- Bedrooms = 3
- Floors = 2
- Age = 10 years

Input vector:

$$
\vec{x} = [1500, 3, 2, 10]
$$

Using the model:

$$
f_{\vec{w},b}(\vec{x}) =
0.1(1500) + 4(3) + 10(2) - 2(10) + 80
$$

$$
= 150 + 12 + 20 - 20 + 80
$$

$$
= 242
$$

Since the output is in thousands of dollars:

$$
\boxed{\text{Predicted House Price} = \$242,000}
$$

## 4. General Formula for n Features:

For any number of features, the multiple linear regression model can be written as:

$$
f_{\vec{w},b}(\vec{x}) =
w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

Where:

- $n$ = Total number of features.
- $x_j$ = Value of the $j$-th feature.
- $w_j$ = Weight corresponding to the $j$-th feature.
- $b$ = Bias.

This formula works for any number of input features.

## 5. Weight Vector ($\vec{w}$):

Instead of writing all weights separately, we can combine them into a single vector called the **Weight Vector**.

$$
\vec{w} = [w_1, w_2, w_3, \ldots, w_n]
$$

For our house price example:

$$
\vec{w} = [0.1, 4, 10, -2]
$$

The weight vector contains all the model weights.

## 6. Feature Vector ($\vec{x}$):

Similarly, all input features can be combined into a single vector called the **Feature Vector**.

$$
\vec{x} = [x_1, x_2, x_3, \ldots, x_n]
$$

For our numerical example:

$$
\vec{x} = [1500, 3, 2, 10]
$$

Therefore:

- $\vec{w}$ contains the model weights.
- $\vec{x}$ contains the input features.
- $b$ is a single numerical value (not a vector).

The weight vector and bias together represent the model parameters.

## 7. Dot Product:

The dot product is a mathematical operation from linear algebra performed between two vectors.

To calculate the dot product:

1. Multiply corresponding elements of both vectors.
2. Add all the resulting products.

Suppose:

$$
\vec{w} = [w_1, w_2, w_3]
$$

$$
\vec{x} = [x_1, x_2, x_3]
$$

Their dot product is:

$$
\vec{w} \cdot \vec{x}
= w_1x_1 + w_2x_2 + w_3x_3
$$

For $n$ features:

$$
\vec{w} \cdot \vec{x}
= \sum_{j=1}^{n} w_jx_j
$$

### Numerical Example:

Given:

$$
\vec{w} = [0.1, 4, 10, -2]
$$

$$
\vec{x} = [1500, 3, 2, 10]
$$

Calculate the dot product:

$$
\vec{w} \cdot \vec{x}
= 0.1(1500) + 4(3) + 10(2) + (-2)(10)
$$

$$
= 150 + 12 + 20 - 20
$$

$$
= 162
$$

Now add the bias:

$$
f_{\vec{w},b}(\vec{x}) = 162 + 80
$$

$$
= 242
$$

Therefore, the predicted house price is $242,000.

## 8. Simplified Multiple Linear Regression Model:

Using the dot product, we can rewrite the multiple linear regression model in a shorter form.

### Expanded Form:

$$
f_{\vec{w},b}(\vec{x}) =
w_1x_1 + w_2x_2 + \cdots + w_nx_n + b
$$

### Dot Product Form:

$$
\boxed{f_{\vec{w},b}(\vec{x}) = \vec{w} \cdot \vec{x} + b}
$$

Both expressions are mathematically equivalent and produce exactly the same prediction.

The dot product notation makes the model shorter and easier to work with, especially when the dataset contains many features.

## 9. Univariate vs. Multiple Linear Regression:

| Univariate Linear Regression | Multiple Linear Regression |
|------------------------------|----------------------------|
| Uses one input feature. | Uses multiple input features. |
| Has one weight ($w$). | Has multiple weights ($w_1, w_2, \ldots, w_n$). |
| Model: $wx+b$ | Model: $\vec{w}\cdot\vec{x}+b$ |



# Vectorization:

## 1. Introduction to Vectorization:

> Vectorization is a programming technique that allows us to perform operations on entire vectors or arrays instead of processing individual elements one by one using loops.

In machine learning, we frequently perform mathematical operations on vectors and matrices. Vectorization makes these operations easier to implement and more computationally efficient.

### Benefits of Vectorization:

1. **Shorter Code:** We can perform complex mathematical operations using fewer lines of code.
2. **Faster Execution:** Vectorized operations are generally much faster than Python loops for large numerical computations.
3. **Parallel Processing:** Optimized numerical libraries can take advantage of parallel hardware.
4. **Better Scalability:** Vectorization helps machine learning algorithms work efficiently with large datasets and models.

Vectorization allows us to take advantage of:
- Numerical linear algebra libraries such as NumPy.
- CPU (Central Processing Unit).
- GPU (Graphics Processing Unit).

## 2. NumPy and Array Indexing:

NumPy is one of the most widely used numerical computing libraries in Python and machine learning.

It provides efficient operations for working with vectors, matrices, and arrays.

### Example:

```python
import numpy as np

w = np.array([1, 2, 3])
b = 4
x = np.array([10, 20, 30])
```

Here:
- `w` = Weight vector.
- `b` = Bias.
- `x` = Feature vector.


## 3. Three Ways to Calculate the Prediction:

Suppose we have three features and three corresponding weights.

The mathematical model is:

$$
f_{\vec{w},b}(\vec{x}) = \vec{w}\cdot\vec{x}+b
$$

We can implement this model in three different ways.

### Method 1: Without Vectorization (Manual Calculation):

We can manually multiply each weight by its corresponding feature and add all the products.

```python
f = w[0] * x[0] + w[1] * x[1] + w[2] * x[2] + b
```

This works for a small number of features, but becomes inconvenient when the number of features increases.

For example, writing separate multiplications for 100,000 features would be impractical.

### Method 2: Without Vectorization (Using a For Loop):

Instead of writing every multiplication manually, we can use a for loop.

```python
f = 0

for j in range(0, 3):
    f = f + w[j] * x[j]

f = f + b
```

**Explanation:**

1. Initialize `f = 0` because we need to accumulate the products.
2. The loop iterates through all feature indices.
3. Multiply each weight by its corresponding feature.
4. Add the products to `f`.
5. Add the bias outside the loop.

Mathematically:

$$
f = \sum_{j=1}^{n} w_jx_j+b
$$

**Important Python Concept:**

```python
range(0, 3)
```

Generates:

$$
0,1,2
$$

Python does not include the ending value of the range.

Similarly:

```python
range(3)
```

Also generates 0, 1, and 2.

This method is more flexible than manual calculation, but still uses a loop.

### Method 3: With Vectorization (Using NumPy Dot):

NumPy provides a built-in function called `np.dot()` to calculate the dot product of two vectors.

```python
f = np.dot(w, x) + b
```

This single statement calculates:

$$
\boxed{f = \vec{w}\cdot\vec{x}+b}
$$

The `np.dot()` function multiplies corresponding elements of two vectors and sums their products.

It performs the dot product using optimized numerical computing routines.

This is generally much more efficient than writing manual calculations or Python loops for large arrays.

## 4. Comparison of the Three Methods:

| Method | Implementation | Characteristics |
|--------|----------------|-----------------|
| Manual Calculation | Individual multiplications | Difficult to maintain for many features. |
| For Loop | Iterates through elements | Flexible but involves sequential Python-level looping. |
| Vectorization | `np.dot(w, x)` | Shorter code and generally faster execution. |

Vectorization has two major advantages:

1. It makes code shorter, cleaner, and easier to understand.
2. It makes numerical calculations faster by using optimized implementations and hardware capabilities.



## 5. How Vectorization Works Behind the Scenes:

To understand why vectorization is faster, consider two vectors containing 16 elements each.

$$
\vec{w} = [w_1,w_2,\ldots,w_{16}]
$$

$$
\vec{x} = [x_1,x_2,\ldots,x_{16}]
$$

We want to calculate their dot product:

$$
\vec{w}\cdot\vec{x}
=
\sum_{j=1}^{16}w_jx_j
$$

### Without Vectorization:

When using a for loop, calculations are generally performed one after another.

For example:

```python
f = 0

for j in range(16):
    f += w[j] * x[j]
```

The computer processes the iterations sequentially:

- First, it calculates the product at index 0.
- Then, it calculates the product at index 1.
- Then, it calculates the product at index 2.
- It continues until index 15.

Each loop iteration performs its calculation before the next iteration proceeds.

### With Vectorization:

Using NumPy:

```python
f = np.dot(w, x)
```

The computer can use optimized numerical routines and parallel-processing hardware to perform multiple multiplications efficiently.

Instead of processing every multiplication through a separate Python loop iteration, the hardware can process multiple pairs of values simultaneously.

After multiplication, specialized optimized routines efficiently sum the resulting products.

**Result:** Vectorized implementations can perform calculations in significantly less time than ordinary Python loops, particularly for large numerical operations.

## 6. Vectorization in Gradient Descent:

Vectorization is not limited to calculating predictions. It can also be used to update model parameters during gradient descent.

In multiple linear regression, we may have many weights that need to be updated.

Suppose we have 16 weights:

$$
\vec{w} = [w_1,w_2,\ldots,w_{16}]
$$

We also have 16 corresponding derivative values:

$$
\vec{d} = [d_1,d_2,\ldots,d_{16}]
$$

For this example, Andrew ignores the bias parameter.

### Gradient Descent Update Rule:

The update rule for an individual weight is:

$$
w_j = w_j-\alpha d_j
$$

Where:

- $w_j$ = Current weight.
- $\alpha$ = Learning rate.
- $d_j$ = Derivative of the cost function with respect to the weight $w_j$.

Andrew uses the learning rate:

$$
\alpha = 0.1
$$

Therefore, the update rule becomes:

$$
w_j = w_j-0.1d_j
$$

### A. Without Vectorization:

We need to update each weight individually.

For example:

$$
w_1=w_1-0.1d_1
$$

$$
w_2=w_2-0.1d_2
$$

$$
w_3=w_3-0.1d_3
$$

Continue until:

$$
w_{16}=w_{16}-0.1d_{16}
$$

Instead of writing all updates manually, we can use a for loop.

```python
for j in range(16):
    w[j] = w[j] - 0.1 * d[j]
```

This loop updates all 16 weights one by one.

### B. With Vectorization:

Using vectorization, we can update all weights using a single NumPy statement.

```python
w = w - 0.1 * d
```

Here:
- `w` contains all 16 weights.
- `d` contains all 16 derivative values.
- `0.1` represents the learning rate.

NumPy performs element-wise multiplication and subtraction.

Mathematically:

$$
\boxed{\vec{w}=\vec{w}-0.1\vec{d}}
$$

This is equivalent to performing all 16 individual updates:

$$
\begin{aligned}
w_1 &= w_1-0.1d_1\\
w_2 &= w_2-0.1d_2\\
w_3 &= w_3-0.1d_3\\
&\vdots\\
w_{16} &= w_{16}-0.1d_{16}
\end{aligned}
$$

All element-wise operations can be executed efficiently using optimized array operations.

### Understanding the Derivative Vector:

The derivative vector contains the derivatives corresponding to each weight.

For example:

$$
\vec{d} =
\left[
\frac{\partial J}{\partial w_1},
\frac{\partial J}{\partial w_2},
\ldots,
\frac{\partial J}{\partial w_{16}}
\right]
$$

Where:
- $J$ = Cost function.
- $\frac{\partial J}{\partial w_j}$ = Derivative of the cost function with respect to the $j$-th weight.

These derivatives indicate how each weight should be adjusted to reduce the cost function.

The calculation of these derivatives will be covered in the upcoming gradient descent lectures.

## 7. Importance of Vectorization for Large Datasets:

Vectorization becomes increasingly important when working with large machine learning problems.

Consider:
- Thousands of features.
- Millions of training examples.
- Large numbers of model parameters.
- Repeated gradient descent iterations.

Without vectorization, calculations using ordinary Python loops can become very slow.

With vectorization, we can use optimized numerical operations and parallel hardware to perform computations more efficiently.

For large machine learning problems, this can make a significant difference in execution time.

It can be the difference between an algorithm taking minutes and taking many hours, depending on the implementation and problem size.

````python
def predict_single_loop(x, w, b): 
    """
    single predict using linear regression
    
    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters    
      b (scalar):  model parameter     
      
    Returns:
      p (scalar):  prediction
    """
    n = x.shape[0]
    p = 0
    for i in range(n):
        p_i = x[i] * w[i]  
        p = p + p_i         
    p = p + b                
    return p
````

````python
def predict(x, w, b): 
    """
    single predict using linear regression
    Args:
      x (ndarray): Shape (n,) example with multiple features
      w (ndarray): Shape (n,) model parameters   
      b (scalar):             model parameter 
      
    Returns:
      p (scalar):  prediction
    """
    p = np.dot(x, w) + b     
    return p    
````

````python
def compute_cost(X, y, w, b): 
    """
    compute cost
    Args:
      X (ndarray (m,n)): Data, m examples with n features
      y (ndarray (m,)) : target values
      w (ndarray (n,)) : model parameters  
      b (scalar)       : model parameter
      
    Returns:
      cost (scalar): cost
    """
    m = X.shape[0]
    cost = 0.0
    for i in range(m):                                
        f_wb_i = np.dot(X[i], w) + b           #(n,)(n,) = scalar (see np.dot)
        cost = cost + (f_wb_i - y[i])**2       #scalar
    cost = cost / (2 * m)                      #scalar    
    return cost
````

````python
# Compute and display cost using our pre-chosen optimal parameters. 
cost = compute_cost(X_train, y_train, w_init, b_init)
print(f'Cost at optimal w : {cost}')
````

````python
def compute_gradient(x, y, w, b):
    m, n = x.shape   # m gives numbers records while n gives the features used in prediction
    dj_dw = np.zeros((n,))
    dj_db = 0.

    for i in range(m):
        err = (np.dot(x[i], w) + b) - y[i]
        for j in range(n):
            dj_dw[j] = dj_dw[j] + err * x[i, j]
        dj_db = dj_db + err                        
    dj_dw = dj_dw / m                                
    dj_db = dj_db / m                                
        
    return dj_db, dj_dw
````

````python
def gradient_descent(X, y, w_in, b_in, cost_function, gradient_function, alpha, num_iters): 
    """
    Performs batch gradient descent to learn w and b. Updates w and b by taking 
    num_iters gradient steps with learning rate alpha
    
    Args:
      X (ndarray (m,n))   : Data, m examples with n features
      y (ndarray (m,))    : target values
      w_in (ndarray (n,)) : initial model parameters  
      b_in (scalar)       : initial model parameter
      cost_function       : function to compute cost
      gradient_function   : function to compute the gradient
      alpha (float)       : Learning rate
      num_iters (int)     : number of iterations to run gradient descent
      
    Returns:
      w (ndarray (n,)) : Updated values of parameters 
      b (scalar)       : Updated value of parameter 
      """
    
    # An array to store cost J and w's at each iteration primarily for graphing later
    J_history = []
    w = copy.deepcopy(w_in)  #avoid modifying global w within function
    b = b_in
    
    for i in range(num_iters):

        # Calculate the gradient and update the parameters
        dj_db,dj_dw = gradient_function(X, y, w, b)   

        # Update Parameters using w, b, alpha and gradient
        w = w - alpha * dj_dw               
        b = b - alpha * dj_db               
      
        # Save cost J at each iteration
        if i<100000:      # prevent resource exhaustion 
            J_history.append( cost_function(X, y, w, b))

        # Print cost every at intervals 10 times or as many iterations if < 10
        if i% math.ceil(num_iters / 10) == 0:
            print(f"Iteration {i:4d}: Cost {J_history[-1]:8.2f}   ")
        
    return w, b, J_history #return final w,b and J history for graphing
````
