# **Feature Scaling:**

## 1. Introduction to Feature Scaling:

Feature Scaling is a technique used to transform the values of different input features so that they have comparable numerical ranges.

In machine learning, different features may have very different ranges of values.

For example, when predicting house prices:

| Feature | Description | Range |
|---------|-------------|-------|
| $x_1$ | House size (square feet) | 300–2000 |
| $x_2$ | Number of bedrooms | 0–5 |

Notice that house size has much larger numerical values than the number of bedrooms.

Having features with very different scales can cause Gradient Descent to run slowly.

Feature Scaling helps solve this problem by transforming the features into comparable ranges.

### Main Benefits of Feature Scaling:

1. Helps Gradient Descent converge faster.
2. Makes the cost function contours more balanced.
3. Reduces unnecessary zigzagging during Gradient Descent.
4. Makes training more computationally efficient.

**Important:** Feature Scaling changes the numerical representation of input features, not the actual information they represent.

## 2. Relationship Between Features and Their Weights:

Recall the Multiple Linear Regression model:

$$
f_{\vec{w},b}(\vec{x})=w_1x_1+w_2x_2+b
$$

Each feature has its own corresponding weight.

Consider:

$$
x_1=2000
$$

$$
x_2=5
$$

Where:
- $x_1$ = House size.
- $x_2$ = Number of bedrooms.

Andrew Ng gives an example where the predicted price is measured in thousands of dollars.

Consider the parameters:

$$
w_1=0.1,\quad w_2=50,\quad b=50
$$

Substituting into the model:

$$
f(x)=0.1(2000)+50(5)+50
$$

$$
=200+250+50
$$

$$
=500
$$

Therefore:

$$
\boxed{\text{Predicted Price}=\$500,000}
$$

### Understanding the Weights:

| Feature | Maximum Value | Weight |
|---------|---------------|--------|
| House size | 2000 | 0.1 |
| Bedrooms | 5 | 50 |

A feature with a larger numerical range often has a relatively smaller weight, while a feature with a smaller numerical range may have a relatively larger weight.

This happens because each feature is multiplied by its corresponding weight when calculating the prediction.

For example:

$$
0.1\times2000=200
$$

$$
50\times5=250
$$

Both features contribute comparable amounts to the prediction.

However, this is an intuition rather than a fixed rule about the weights a model must learn.

### What Happens With Inappropriate Weights?

Suppose we reverse the weights:

$$
w_1=50,\quad w_2=0.1,\quad b=50
$$

Then:

$$
f(x)=50(2000)+0.1(5)+50
$$

$$
=100000+0.5+50
$$

$$
=100050.5
$$

Since the output is in thousands of dollars:

$$
\text{Predicted Price}=\$100,050,500
$$

This is far from the actual price of $500,000.

**Important:** We do not manually choose the correct weights. Gradient Descent learns them. However, different feature scales can make the optimization process much slower.

## 3. Why Do Different Feature Scales Affect Gradient Descent?

Gradient Descent is an optimization algorithm that repeatedly updates model parameters to minimize the cost function.

Its update rule is:

$$
w_j=w_j-\alpha\frac{\partial J}{\partial w_j}
$$

Where:

- $w_j$ = Current model parameter.
- $\alpha$ = Learning rate.
- $J$ = Cost function.
- $\frac{\partial J}{\partial w_j}$ = Derivative of the cost function with respect to the parameter.

When features have very different numerical ranges, the cost function can become much more sensitive to changes in one parameter than another.

For example:

- House size ranges from 300 to 2000.
- Bedrooms range from 0 to 5.

A small change in $w_1$ can have a large impact on predictions because it is multiplied by relatively large house-size values.

In comparison, it may take a larger change in $w_2$ to create a similar effect because bedroom values are much smaller.

### Understanding Contour Plots:

A contour plot is a graphical representation of the cost function.

- The horizontal axis represents $w_1$.
- The vertical axis represents $w_2$.
- The center represents the minimum cost.
- Each contour represents a different cost level.

When features have very different scales, the contours can become long and narrow, forming elongated ellipses.

### Why Does This Cause Problems?

When Gradient Descent operates on an elongated cost function:

1. It takes a step toward the minimum.
2. It may move too far across the narrow direction.
3. It then corrects its direction.
4. It may move too far in the opposite direction.
5. This process creates a zigzagging path.

As a result, Gradient Descent may take many iterations to approach the minimum.

### How Feature Scaling Helps:

Feature Scaling transforms the features into comparable numerical ranges.

This can make the cost function contours more circular rather than long and narrow.

Gradient Descent can then follow a more direct path toward the minimum.

**Conclusion:** Different feature scales can make Gradient Descent slow. Feature Scaling helps improve the efficiency of Gradient Descent.

## 4. Techniques of Feature Scaling:
The following are feature scaling techniques:
1. Scaling by Dividing by the Maximum.
2. Mean Normalization.
3. Z-Score Normalization.

Consider the following original feature ranges:

| Feature | Minimum | Maximum |
|---------|---------|---------|
| $x_1$ (House size) | 300 | 2000 |
| $x_2$ (Bedrooms) | 0 | 5 |

We will apply all three techniques to these features.

## 5. Technique 1: Dividing by the Maximum:

In this technique, each feature value is divided by the maximum value of that feature.

### Formula:

$$
\boxed{x_{\text{scaled}}=\frac{x}{x_{\max}}}
$$

Where:

- $x$ = Original feature value.
- $x_{\max}$ = Maximum value of that feature.

### Feature 1: House Size

Given:

$$
x_{1,\min}=300
$$

$$
x_{1,\max}=2000
$$

The formula becomes:

$$
x_{1,\text{scaled}}=\frac{x_1}{2000}
$$

Calculate the minimum:

$$
\frac{300}{2000}=0.15
$$

Calculate the maximum:

$$
\frac{2000}{2000}=1
$$

New range:

$$
\boxed{0.15\leq x_{1,\text{scaled}}\leq1}
$$

### Feature 2: Number of Bedrooms

Given:

$$
x_{2,\min}=0
$$

$$
x_{2,\max}=5
$$

The formula becomes:

$$
x_{2,\text{scaled}}=\frac{x_2}{5}
$$

Calculate the minimum:

$$
\frac{0}{5}=0
$$

Calculate the maximum:

$$
\frac{5}{5}=1
$$

New range:

$$
\boxed{0\leq x_{2,\text{scaled}}\leq1}
$$

### Comparison

| Feature | Original Range | Scaled Range |
|---------|----------------|--------------|
| House size | 300–2000 | 0.15–1 |
| Bedrooms | 0–5 | 0–1 |

Both features now have comparable numerical ranges.

### Numerical Example

Suppose a house has:

$$
x_1=1200,\quad x_2=3
$$

House size:

$$
x_{1,\text{scaled}}=\frac{1200}{2000}=0.6
$$

Bedrooms:

$$
x_{2,\text{scaled}}=\frac{3}{5}=0.6
$$

New feature vector:

$$
\boxed{\vec{x}_{\text{scaled}}=[0.6,0.6]}
$$

**Important:** This technique divides by the maximum. It is not the same as min-max scaling, which uses:

$$
x_{\text{min-max}}=
\frac{x-x_{\min}}{x_{\max}-x_{\min}}
$$

## 6. Technique 2: Mean Normalization:

Mean Normalization transforms feature values so that they are centered around zero.

It involves two operations:

1. Subtract the mean from each feature value.
2. Divide by the difference between the maximum and minimum values.

### What Is Mean?

Mean is the average of a collection of numbers.

### Formula for Mean:

$$
\boxed{\mu=\frac{\text{Sum of all values}}{\text{Number of values}}}
$$

Where $\mu$ (Greek letter mu) represents the mean.

### Example of Mean Calculation:

Suppose we have five house sizes:

$$
[400,500,600,700,800]
$$

Calculate the mean:

$$
\mu=\frac{400+500+600+700+800}{5}
$$

$$
\mu=\frac{3000}{5}
$$

$$
\boxed{\mu=600}
$$

Therefore, the average house size is 600 square feet.

### Mean Normalization Formula

$$
\boxed{
x_{\text{normalized}}=
\frac{x-\mu}{x_{\max}-x_{\min}}
}
$$

Where:

- $x$ = Original feature value.
- $\mu$ = Mean of the feature.
- $x_{\max}$ = Maximum value of the feature.
- $x_{\min}$ = Minimum value of the feature.

Subtracting the mean centers the values around zero, while dividing by the range adjusts their numerical scale.

### Feature 1: House Size

Given:

$$
x_{1,\min}=300
$$

$$
x_{1,\max}=2000
$$

$$
\mu_1=600
$$

First, calculate the range:

$$
2000-300=1700
$$

Apply the mean normalization formula:

$$
x_{1,\text{normalized}}=
\frac{x_1-600}{1700}
$$

Calculate the minimum:

$$
\frac{300-600}{1700}
$$

$$
=\frac{-300}{1700}
$$

$$
\boxed{-0.18}
$$

Calculate the maximum:

$$
\frac{2000-600}{1700}
$$

$$
=\frac{1400}{1700}
$$

$$
\boxed{0.82}
$$

New range:

$$
\boxed{-0.18\text{ to }0.82}
$$

### Feature 2: Number of Bedrooms

Given:

$$
x_{2,\min}=0
$$

$$
x_{2,\max}=5
$$

$$
\mu_2=2.3
$$

First, calculate the range:

$$
5-0=5
$$

Apply the formula:

$$
x_{2,\text{normalized}}=
\frac{x_2-2.3}{5}
$$

Calculate the minimum:

$$
\frac{0-2.3}{5}
$$

$$
=\frac{-2.3}{5}
$$

$$
\boxed{-0.46}
$$

Calculate the maximum:

$$
\frac{5-2.3}{5}
$$

$$
=\frac{2.7}{5}
$$

$$
\boxed{0.54}
$$

New range:

$$
\boxed{-0.46\text{ to }0.54}
$$

### Comparison

| Feature | Original Range | Mean | Mean Normalized Range |
|---------|----------------|------|-----------------------|
| House size | 300–2000 | 600 | -0.18 to 0.82 |
| Bedrooms | 0–5 | 2.3 | -0.46 to 0.54 |

Both features are now centered around zero and have comparable numerical ranges.

### Numerical Example:

Consider a house with:

$$
x_1=1200,\quad x_2=3
$$

For house size:

$$
x_{1,\text{normalized}}=
\frac{1200-600}{1700}
$$

$$
=\frac{600}{1700}
$$

$$
=0.353
$$

For bedrooms:

$$
x_{2,\text{normalized}}=
\frac{3-2.3}{5}
$$

$$
=\frac{0.7}{5}
$$

$$
=0.14
$$

New feature vector:

$$
\boxed{\vec{x}_{\text{normalized}}=[0.353,0.14]}
$$

## 7. Technique 3: Z-Score Normalization:

Z-Score Normalization is another feature-scaling technique.

It uses two statistical values:

1. Mean ($\mu$).
2. Standard Deviation ($\sigma$).

### What Is Standard Deviation?

Standard Deviation measures how spread out the values are around their mean.

- A small standard deviation indicates that values are relatively close to the mean.
- A large standard deviation indicates that values are more spread out.

### Formula for Population Standard Deviation:

$$
\boxed{
\sigma=
\sqrt{
\frac{1}{m}
\sum_{i=1}^{m}(x^{(i)}-\mu)^2
}
}
$$

Where:

- $\sigma$ = Standard deviation.
- $m$ = Number of observations.
- $x^{(i)}$ = Individual observation.
- $\mu$ = Mean.

### Example of Standard Deviation Calculation:

Suppose our dataset is:

$$
[2,4,6]
$$

**Step 1: Calculate the mean.**

$$
\mu=\frac{2+4+6}{3}=4
$$

**Step 2: Subtract the mean from every value.**

| Value | Calculation | Difference |
|-------|-------------|------------|
| 2 | 2 - 4 | -2 |
| 4 | 4 - 4 | 0 |
| 6 | 6 - 4 | 2 |

**Step 3: Square every difference.**

| Difference | Squared Difference |
|------------|-------------------|
| -2 | 4 |
| 0 | 0 |
| 2 | 4 |

**Step 4: Calculate the average of squared differences.**

$$
\frac{4+0+4}{3}
$$

$$
=\frac{8}{3}
$$

**Step 5: Take the square root.**

$$
\sigma=\sqrt{\frac{8}{3}}
$$

$$
\boxed{\sigma\approx1.633}
$$

This is the population standard deviation.

### Important Statistical Difference:

Population standard deviation divides by $m$:

$$
\sigma=
\sqrt{\frac{\sum(x^{(i)}-\mu)^2}{m}}
$$

Sample standard deviation usually divides by $m-1$:

$$
s=
\sqrt{\frac{\sum(x^{(i)}-\mu)^2}{m-1}}
$$

For this course's example, we use population-style standard deviation.

### Z-Score Normalization Formula:

Z-Score Normalization subtracts the mean and divides the result by the standard deviation.

$$
\boxed{
x_{\text{z-score}}=\frac{x-\mu}{\sigma}
}
$$

Where:

- $x$ = Original feature value.
- $\mu$ = Mean of the feature.
- $\sigma$ = Standard deviation of the feature.

### Feature 1: House Size

Andrew assumes:

$$
\mu_1=600
$$

$$
\sigma_1=450
$$

Apply the formula:

$$
x_{1,\text{z-score}}=
\frac{x_1-600}{450}
$$

Calculate the minimum:

$$
\frac{300-600}{450}
$$

$$
=\frac{-300}{450}
$$

$$
\boxed{-0.67}
$$

Calculate the maximum:

$$
\frac{2000-600}{450}
$$

$$
=\frac{1400}{450}
$$

$$
\boxed{3.11}
$$

New range:

$$
\boxed{-0.67\text{ to }3.11}
$$

### Feature 2: Number of Bedrooms

Andrew assumes:

$$
\mu_2=2.3
$$

$$
\sigma_2=1.4
$$

Apply the formula:

$$
x_{2,\text{z-score}}=
\frac{x_2-2.3}{1.4}
$$

Calculate the minimum:

$$
\frac{0-2.3}{1.4}
$$

$$
\boxed{-1.64}
$$

Calculate the maximum:

$$
\frac{5-2.3}{1.4}
$$

$$
\boxed{1.93}
$$

New range:

$$
\boxed{-1.64\text{ to }1.93}
$$

### Comparison:

| Feature | Original Range | Mean | Standard Deviation | Z-Score Range |
|---------|----------------|------|-------------------|---------------|
| House size | 300–2000 | 600 | 450 | -0.67 to 3.11 |
| Bedrooms | 0–5 | 2.3 | 1.4 | -1.64 to 1.93 |

Both features now have comparable numerical ranges.

### Numerical Example:

Consider a house with:

$$
x_1=1200,\quad x_2=3
$$

For house size:

$$
x_{1,\text{z-score}}=
\frac{1200-600}{450}
$$

$$
=\frac{600}{450}
$$

$$
=1.33
$$

For bedrooms:

$$
x_{2,\text{z-score}}=
\frac{3-2.3}{1.4}
$$

$$
=\frac{0.7}{1.4}
$$

$$
=0.5
$$

New feature vector:

$$
\boxed{\vec{x}_{\text{z-score}}=[1.33,0.5]}
$$

## 8. Comparison of All Three Feature Scaling Techniques:

| Technique | Formula | Main Operation |
|-----------|---------|---------------|
| Dividing by Maximum | $\frac{x}{x_{\max}}$ | Divide by the maximum value. |
| Mean Normalization | $\frac{x-\mu}{x_{\max}-x_{\min}}$ | Subtract the mean and divide by the range. |
| Z-Score Normalization | $\frac{x-\mu}{\sigma}$ | Subtract the mean and divide by standard deviation. |

### Results for the Same House:

For the house:

$$
\vec{x}=[1200,3]
$$

The transformed vectors are:

**Dividing by Maximum:**

$$
[0.6,0.6]
$$

**Mean Normalization:**

$$
[0.353,0.14]
$$

**Z-Score Normalization:**

$$
[1.33,0.5]
$$

All three methods produce different numerical values because they use different mathematical transformations.

Their common objective is to make feature scales comparable.

### Important Examples

**Feature 1:**

If its values range from -100 to +100, its numerical scale is much larger than the desired approximate range.

It is probably better to scale this feature.

**Feature 2:**

If its values range from -0.001 to +0.001, its values are extremely small compared with other features.

Scaling may also help.

**Feature 3:**

If its values range from 98.6 to 105, such as patient body temperature in Fahrenheit, scaling can help because these values are relatively large compared with other scaled features.

**Conclusion:** When feature ranges are very different, feature scaling can improve Gradient Descent. When in doubt, applying suitable feature scaling is often a good choice.

## 10. Practical Implementation Using NumPy

NumPy provides functions for calculating maximum values, minimum values, mean and standard deviation.

Consider a training dataset:

```python
import numpy as np

X_train = np.array([
    [1000, 2],
    [1200, 3],
    [1500, 3],
    [1800, 4],
    [2000, 5]
])
```

Each row represents one training example, and each column represents one feature.

### A. Scaling by Maximum

```python
X_scaled = X_train / np.max(X_train, axis=0)
```

### B. Mean Normalization

```python
mu = np.mean(X_train, axis=0)

feature_range = (
    np.max(X_train, axis=0) -
    np.min(X_train, axis=0)
)

X_norm = (X_train - mu) / feature_range
```

### C. Z-Score Normalization

```python
mu = np.mean(X_train, axis=0)

sigma = np.std(X_train, axis=0)

X_zscore = (X_train - mu) / sigma
```

**Understanding `axis=0`:**

In a NumPy array, `axis=0` tells NumPy to calculate the result separately for each column.

Therefore:
- The first column gets its own maximum, mean and standard deviation.
- The second column gets its own maximum, mean and standard deviation.

### Scaling New Data

Once we calculate the statistics from the training dataset, we should reuse them for test data and future predictions.

For example:

```python
X_test_zscore = (X_test - mu) / sigma
```

We should not calculate a new mean and standard deviation from the test dataset.

This maintains consistency and prevents data leakage.
