# **Regularization:**

## 1. Introduction:

Regularization is a technique used to reduce Overfitting by encouraging a machine learning model to keep its parameter values relatively small.

In the previous topic, we learned that Overfitting occurs when a model fits the training data too closely but performs poorly on new, unseen examples.

For example, consider a Polynomial Regression model:

$$
f(x)=w_1x+w_2x^2+w_3x^3+w_4x^4+b
$$

If the model learns unnecessarily large parameter values, it may produce an overly complicated curve that fits the training data closely but does not generalize well.

Regularization helps control this problem by penalizing large weights.

### Main Idea of Regularization:

Instead of completely removing features, Regularization encourages the model to reduce the values of its weights.

For example:

Original weights:

$$
\vec{w}=[10,20,100,200]
$$

After applying regularization during training, the weights might become:

$$
\vec{w}=[8,12,5,2]
$$

These are illustrative values.

Notice that all features are still present, but their weights have become smaller.

### Why Does Regularization Help?

1. It discourages unnecessarily large parameter values.
2. It reduces the influence of individual features.
3. It helps control model complexity.
4. It can reduce Overfitting.
5. It can improve performance on unseen examples.

**Important:** Regularization does not necessarily remove features. It encourages their weights to become smaller.

## 2. Understanding Regularization Through Feature Removal:

Consider the polynomial model:

$$
f(x)=w_1x+w_2x^2+w_3x^3+w_4x^4+b
$$

Suppose we want to completely remove the contribution of the features $x^3$ and $x^4$.

We can set:

$$
w_3=0,\quad w_4=0
$$

The model becomes:

$$
f(x)=w_1x+w_2x^2+b
$$

This is equivalent to removing those two features.

However, Regularization takes a different approach.

Instead of setting the weights exactly to zero, it encourages them to become smaller.

For example:

**Feature removal:**

$$
\vec{w}=[10,20,0,0]
$$

**Regularization:**

$$
\vec{w}=[8,12,5,2]
$$

Regularization keeps all features but reduces their influence.

## 3. Original Cost Function of Linear Regression:

Recall the original Squared Error Cost Function:

$$
\boxed{
J(\vec{w},b)=
\frac{1}{2m}
\sum_{i=1}^{m}
\left(
f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)}
\right)^2
}
$$

Where:

- $m$ = Number of training examples.
- $n$ = Number of features.
- $f_{\vec{w},b}(\vec{x}^{(i)})$ = Predicted output.
- $y^{(i)}$ = Actual output.
- $\vec{w}$ = Weight vector.
- $b$ = Bias.

This Cost Function measures how well the model predicts the training data.

However, it does not directly penalize large parameter values.

To reduce Overfitting, we modify this Cost Function by adding a Regularization term.

## 4. Introducing the Regularization Term:

Suppose we have a polynomial model:

$$
f(x)=w_1x+w_2x^2+w_3x^3+w_4x^4+b
$$

Initially, we want to penalize only $w_3$ and $w_4$.

We can add:

$$
1000w_3^2+1000w_4^2
$$

to the original Cost Function.

The modified Cost Function becomes:

$$
J_{\text{new}}=
J_{\text{original}}+1000w_3^2+1000w_4^2
$$

Since the algorithm wants to minimize the new Cost Function, it will try to keep $w_3$ and $w_4$ small.

This reduces the influence of the corresponding features $x^3$ and $x^4$.

### General Regularization Term:

In practice, we may have many features and may not know which weights should be penalized.

Therefore, we usually penalize all the weights.

The general penalty is:

$$
\boxed{\lambda\sum_{j=1}^{n}w_j^2}
$$

Where:

- $\lambda$ = Regularization parameter.
- $n$ = Total number of features.
- $w_j$ = Weight of the $j$-th feature.

## 5. Regularized Cost Function of Linear Regression:

The regularized Cost Function combines two terms:

1. Original Squared Error Cost Function.
2. Regularization term.

### Formula:

$$
\boxed{
J_{\text{reg}}(\vec{w},b)=
\frac{1}{2m}
\sum_{i=1}^{m}
\left(
f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)}
\right)^2
+
\frac{\lambda}{2m}
\sum_{j=1}^{n}w_j^2
}
$$

### Understanding the Two Terms:

**First Term: Original Cost Function**

$$
\frac{1}{2m}
\sum_{i=1}^{m}
\left(
f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)}
\right)^2
$$

This measures the prediction error across the training dataset.

The algorithm tries to minimize this term to make accurate predictions.

**Second Term: Regularization Term**

$$
\frac{\lambda}{2m}
\sum_{j=1}^{n}w_j^2
$$

This penalizes large weights.

The algorithm tries to minimize this term to keep the model parameters small.

### Why Do We Divide Lambda by 2m?

The Regularization term could initially be written as:

$$
\lambda\sum_{j=1}^{n}w_j^2
$$

However, the convention used here is:

$$
\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
$$

This scaling makes both terms use the same $1/(2m)$ convention and simplifies the derivative calculations.

It also helps maintain more consistent regularization behavior when the training-set size changes.

**Important:** Always use the correct derivative corresponding to the exact Cost Function being used.

## 6. Regularization Parameter (Lambda):

Lambda is the regularization parameter that controls how strongly large weights are penalized.

It determines the trade-off between:

1. Fitting the training data well.
2. Keeping the model weights small.

The value of Lambda affects the complexity of the model.

### Case 1: Lambda = 0

If:

$$
\lambda=0
$$

Then:

$$
\frac{\lambda}{2m}
\sum_{j=1}^{n}w_j^2=0
$$

The Regularization term disappears.

Therefore:

$$
J_{\text{reg}}=J_{\text{original}}
$$

No Regularization is applied.

The model may learn unnecessarily large weights and overfit the training data.

Result:

$$
\boxed{\text{Possible Overfitting}}
$$

### Case 2: Lambda Is Extremely Large

Suppose:

$$
\lambda=10^{10}
$$

The Regularization term becomes extremely important.

To minimize the Cost Function, the algorithm tries to make all weights close to zero:

$$
w_1,w_2,\ldots,w_n\approx0
$$

The model becomes approximately:

$$
f(x)\approx b
$$

Since the feature contributions become very small, the model may produce an almost horizontal line in the original one-feature regression example.

The model becomes too simple to capture the actual relationship.

Result:

$$
\boxed{\text{Underfitting}}
$$

### Case 3: An Appropriate Lambda Value

When Lambda is neither too small nor too large, the algorithm balances both objectives.

- It fits the training data reasonably well.
- It keeps parameter values controlled.
- It reduces unnecessary model complexity.
- It can improve generalization.

### Comparison:

| Lambda Value | Effect | Result |
|---|---|---|
| 0 | No Regularization. | May overfit. |
| Very small | Weak penalty. | May still overfit. |
| Appropriate | Balanced penalty. | Better generalization. |
| Very large | Strong penalty. | May underfit. |

**Important:** Lambda is a hyperparameter. It is selected during model development rather than learned as one of the model weights.

## 7. Why Don't We Regularize the Bias?

Our model contains:

$$
w_1,w_2,\ldots,w_n,b
$$

By convention, Regularization is applied to the weights:

$$
\boxed{w_1,w_2,\ldots,w_n}
$$

The bias parameter $b$ is usually not regularized.

Therefore, the Regularization term is:

$$
\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
$$

It does not include:

$$
b^2
$$

The bias represents the intercept of the model. Shrinking it is generally unnecessary for controlling the influence of the input features.

**Important:** In standard implementations, we usually regularize the weights but not the bias.

# Regularized Linear Regression:

## 8. Original Gradient Descent for Linear Regression:

Recall the original Gradient Descent update rules.

### Weight Update:

$$
\boxed{
w_j:=w_j-\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)})
x_j^{(i)}
\right]
}
$$

### Bias Update:

$$
\boxed{
b:=b-\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)})
\right]
}
$$

Where:

- $\alpha$ = Learning rate.
- $m$ = Number of training examples.
- $f(\vec{x}^{(i)})$ = Predicted output.
- $y^{(i)}$ = Actual output.
- $x_j^{(i)}$ = $j$-th feature of training example $i$.

## 9. Gradient Descent for Regularized Linear Regression:

When we add the Regularization term to the original Cost Function, the derivative with respect to each weight receives an additional term.

### Derivative with Respect to Weight:

The original derivative is:

$$
\frac{\partial J}{\partial w_j}
=
\frac{1}{m}
\sum_{i=1}^{m}
(f(\vec{x}^{(i)})-y^{(i)})x_j^{(i)}
$$

The additional derivative from the Regularization term is:

$$
\frac{\lambda}{m}w_j
$$

Therefore:

$$
\boxed{
\frac{\partial J_{\text{reg}}}{\partial w_j}
=
\frac{1}{m}
\sum_{i=1}^{m}
(f(\vec{x}^{(i)})-y^{(i)})x_j^{(i)}
+
\frac{\lambda}{m}w_j
}
$$

### Derivative with Respect to Bias:

Since the bias is not regularized, its derivative remains unchanged:

$$
\boxed{
\frac{\partial J_{\text{reg}}}{\partial b}
=
\frac{1}{m}
\sum_{i=1}^{m}
(f(\vec{x}^{(i)})-y^{(i)})
}
$$

### Final Gradient Descent Update Rules:

**Updating the weights:**

$$
\boxed{
w_j:=w_j-\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(f(\vec{x}^{(i)})-y^{(i)})x_j^{(i)}
+
\frac{\lambda}{m}w_j
\right]
}
$$

**Updating the bias:**

$$
\boxed{
b:=b-\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(f(\vec{x}^{(i)})-y^{(i)})
\right]
}
$$

**Important Observation:**

- The weight update receives an additional Regularization term.
- The bias update remains the same as ordinary Linear Regression.
- All parameters must be updated simultaneously.

## 10. Derivation of the Regularization Gradient:

The Regularization term is:

$$
J_{\text{reg-term}}=
\frac{\lambda}{2m}
\sum_{k=1}^{n}w_k^2
$$

To calculate its derivative with respect to $w_j$:

$$
\frac{\partial}{\partial w_j}
\left(
\frac{\lambda}{2m}
\sum_{k=1}^{n}w_k^2
\right)
$$

Only $w_j^2$ changes when differentiating with respect to $w_j$. All other weights are treated as constants.

Therefore:

$$
=
\frac{\lambda}{2m}
(2w_j)
$$

Cancel the 2:

$$
\boxed{
\frac{\partial J_{\text{reg-term}}}{\partial w_j}
=
\frac{\lambda}{m}w_j
}
$$

For the bias:

$$
\boxed{
\frac{\partial J_{\text{reg-term}}}{\partial b}=0
}
$$

Because $b$ is not part of the Regularization term.

## 11. Understanding Weight Shrinkage:

The weight update for Regularized Linear Regression is:

$$
w_j:=w_j-\alpha
\left[
\frac{1}{m}\sum_{i=1}^{m}(f_i-y^{(i)})x_j^{(i)}
+
\frac{\lambda}{m}w_j
\right]
$$

Rearranging the terms:

$$
w_j:=
w_j-\frac{\alpha\lambda}{m}w_j
-
\frac{\alpha}{m}
\sum_{i=1}^{m}(f_i-y^{(i)})x_j^{(i)}
$$

Taking $w_j$ as a common factor:

$$
\boxed{
w_j:=
\left(1-\frac{\alpha\lambda}{m}\right)w_j
-
\frac{\alpha}{m}
\sum_{i=1}^{m}(f_i-y^{(i)})x_j^{(i)}
}
$$

The first part:

$$
\left(1-\frac{\alpha\lambda}{m}\right)w_j
$$

shrinks the existing weight by multiplying it with a factor that is normally slightly smaller than 1.

The second part performs the usual Gradient Descent correction based on training errors.

### Numerical Example:

Suppose:

$$
\alpha=0.01
$$

$$
\lambda=1
$$

$$
m=50
$$

Calculate the shrinkage factor:

$$
1-\frac{\alpha\lambda}{m}
$$

$$
=1-\frac{0.01(1)}{50}
$$

$$
=1-0.0002
$$

$$
\boxed{0.9998}
$$

If the current weight is:

$$
w_j=100
$$

The shrinkage part gives:

$$
100\times0.9998
$$

$$
\boxed{99.98}
$$

The normal Gradient Descent correction is then applied.

Therefore, Regularization slightly reduces the weight during every iteration, in addition to the normal update.

## 12. Complete Numerical Example of Regularized Gradient Descent:

Suppose:

$$
w_j=10
$$

$$
\alpha=0.01
$$

$$
\lambda=1
$$

$$
m=50
$$

Assume the ordinary data-based gradient is:

$$
\frac{1}{m}\sum_{i=1}^{m}(f_i-y^{(i)})x_j^{(i)}=2
$$

The Regularization gradient is:

$$
\frac{\lambda}{m}w_j
=
\frac{1}{50}(10)
$$

$$
=0.2
$$

Total gradient:

$$
2+0.2=2.2
$$

Now update:

$$
w_j^{\text{new}}=10-0.01(2.2)
$$

$$
=10-0.022
$$

$$
\boxed{w_j^{\text{new}}=9.978}
$$

Without Regularization:

$$
w_j^{\text{new}}=10-0.01(2)
$$

$$
\boxed{w_j^{\text{new}}=9.98}
$$

Comparison:

| Method | Updated Weight |
|---|---:|
| Without Regularization | 9.980 |
| With Regularization | 9.978 |

The additional Regularization term causes a small extra reduction in the weight.

## 13. Simultaneous Updates:

During Gradient Descent, all parameters must be updated simultaneously.

For example, suppose our model has:

$$
w_1,w_2,b
$$

We must first calculate all gradients using the original parameter values.

Then we update all the parameters together.

We should not update $w_1$ and then use its newly updated value when calculating the gradient for $w_2$.

This is the same simultaneous-update rule used in ordinary Gradient Descent.

# Regularized Logistic Regression:

## 14. Why Does Logistic Regression Need Regularization?

Logistic Regression can also suffer from Overfitting.

This can happen when using many features, especially high-order polynomial features.

For example:

$$
z=w_1x_1+w_2x_2+w_3x_1^2+w_4x_1x_2+w_5x_2^2
$$

When many features are included, the model may learn an overly complicated decision boundary.

This can cause poor predictions on unseen data.

Regularization helps reduce this problem by encouraging smaller weights.

## 15. Original Logistic Regression Cost Function:

Recall that the Logistic Regression Cost Function is:

$$
\boxed{
J(\vec{w},b)=
\frac{1}{m}
\sum_{i=1}^{m}
L(f_{\vec{w},b}(\vec{x}^{(i)}),y^{(i)})
}
$$

Where the Logistic Regression loss is:

$$
L(f(x),y)=
\begin{cases}
-\log(f(x)) & \text{if }y=1\\
-\log(1-f(x)) & \text{if }y=0
\end{cases}
$$

The model predicts probabilities using the Sigmoid Function:

$$
\boxed{
f_{\vec{w},b}(\vec{x})=
\frac{1}{1+e^{-(\vec{w}\cdot\vec{x}+b)}}
}
$$

This Cost Function measures the average Logistic Loss across the training dataset.

## 16. Regularized Logistic Regression Cost Function:

To apply Regularization, we add the same penalty term used in Regularized Linear Regression.

The Regularization term is:

$$
\frac{\lambda}{2m}
\sum_{j=1}^{n}w_j^2
$$

Therefore, the new Cost Function becomes:

$$
\boxed{
J_{\text{reg}}(\vec{w},b)=
\frac{1}{m}
\sum_{i=1}^{m}
L(f_{\vec{w},b}(\vec{x}^{(i)}),y^{(i)})
+
\frac{\lambda}{2m}
\sum_{j=1}^{n}w_j^2
}
$$

The first term is the original Logistic Regression Cost Function.

The second term penalizes large weights.

### Combined Log Loss Form:

The piecewise Logistic Loss can also be written in a single expression:

$$
L(f(x),y)=
-y\log(f(x))
-(1-y)\log(1-f(x))
$$

Therefore, Regularized Logistic Regression can also be written as:

$$
\boxed{
\begin{aligned}
J_{\text{reg}}(\vec{w},b)
=&-\frac{1}{m}\sum_{i=1}^{m}
\left[
y^{(i)}\log(f_i)
\right.\\
&\left.+(1-y^{(i)})\log(1-f_i)
\right]\\
&+\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
\end{aligned}
}
$$

Where:

$$
f_i=f_{\vec{w},b}(\vec{x}^{(i)})
$$

**Important:** The regularization term is the same as the one used in Regularized Linear Regression. The difference is the original Cost Function.

## 17. Gradient Descent for Regularized Logistic Regression:

We minimize the Regularized Logistic Regression Cost Function using Gradient Descent.

The update equations have the same structure as Regularized Linear Regression.

### Updating the Weights:

$$
\boxed{
w_j:=w_j-\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)})x_j^{(i)}
+
\frac{\lambda}{m}w_j
\right]
}
$$

### Updating the Bias:

$$
\boxed{
b:=b-\alpha
\left[
\frac{1}{m}
\sum_{i=1}^{m}
(f_{\vec{w},b}(\vec{x}^{(i)})-y^{(i)})
\right]
}
$$

Here, the prediction is calculated using the Logistic Regression model:

$$
f_{\vec{w},b}(\vec{x}^{(i)})=
\frac{1}{1+e^{-(\vec{w}\cdot\vec{x}^{(i)}+b)}}
$$

**Important:**

- The weight derivative contains the additional Regularization term.
- The bias derivative remains unchanged.
- All parameter updates must be simultaneous.

## 18. Why Are Regularized Linear and Logistic Regression Updates Similar?

The two algorithms use different prediction functions.

### Linear Regression:

$$
f_{\vec{w},b}(\vec{x})=
\vec{w}\cdot\vec{x}+b
$$

### Logistic Regression:

$$
f_{\vec{w},b}(\vec{x})=
\frac{1}{1+e^{-(\vec{w}\cdot\vec{x}+b)}}
$$

However, both regularized algorithms use the same general Gradient Descent update structure:

$$
\boxed{
w_j:=w_j-\alpha
\left[
\text{Data-based Gradient}
+
\frac{\lambda}{m}w_j
\right]
}
$$

And both use the same bias update structure:

$$
\boxed{
b:=b-\alpha(\text{Average Prediction Error})
}
$$

The main difference is the prediction function $f$:

- Linear Regression uses the linear function.
- Logistic Regression uses the sigmoid function.

The Logistic Loss combined with the sigmoid function produces the same prediction-error gradient form.

## 19. Difference Between Alpha and Lambda:

These are two different parameters used in Regularized Gradient Descent.

| Parameter | Name | Purpose |
|---|---|---|
| $\alpha$ | Learning Rate | Controls the size of each Gradient Descent step. |
| $\lambda$ | Regularization Parameter | Controls how strongly large weights are penalized. |

### Learning Rate:

A larger learning rate means larger update steps.

A smaller learning rate means smaller update steps.

### Regularization Parameter:

A larger Lambda means a stronger penalty on large weights.

A smaller Lambda means a weaker penalty.

**Important:** Alpha and Lambda perform different jobs and should not be confused.

## 20. Comparison of the Four Algorithms:

| Algorithm | Original Cost Function | Regularization |
|---|---|---|
| Linear Regression | Squared Error | Optional |
| Regularized Linear Regression | Squared Error + Penalty | Yes |
| Logistic Regression | Logistic Loss | Optional |
| Regularized Logistic Regression | Logistic Loss + Penalty | Yes |

### Regularized Linear Regression:

$$
f(\vec{x})=\vec{w}\cdot\vec{x}+b
$$

$$
J_{\text{reg}}=
\text{Squared Error Cost}
+
\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
$$

### Regularized Logistic Regression:

$$
f(\vec{x})=
\frac{1}{1+e^{-(\vec{w}\cdot\vec{x}+b)}}
$$

$$
J_{\text{reg}}=
\text{Logistic Cost}
+
\frac{\lambda}{2m}\sum_{j=1}^{n}w_j^2
$$

The Regularization term is identical in both algorithms.
