# **Overfitting:**

## 1. Introduction:

In Machine Learning, our objective is to train a model that learns patterns from training data and performs well on new, unseen examples.

However, sometimes a model fits the training data too closely and fails to make accurate predictions on new data.

This problem is called **Overfitting**.

For example, suppose we are predicting house prices using house sizes.

A highly complicated polynomial model may produce a very wiggly curve that follows almost every training example.

Although the model performs very well on the training dataset, it may make unrealistic predictions for houses it has never seen before.

### Definition of Overfitting:

> Overfitting occurs when a machine learning model fits the training data too closely, including noise and unnecessary variations, but performs poorly on new, unseen examples.

In simple words, the model memorizes the training data instead of learning the general pattern.

### Generalization:

Generalization is the ability of a trained machine learning model to make accurate predictions on new data that it has never seen before.

The objective of Machine Learning is not just to perform well on training data but also to generalize well to unseen data.

## 2. High Variance:

Another term commonly associated with Overfitting is **High Variance**.

A high-variance model is highly sensitive to the particular training examples it receives.

For example, if a complicated polynomial model is trained on a small dataset, adding or removing a few training examples may significantly change the shape of the fitted curve.

This means the model is learning unnecessary variations from the training data.

Therefore:

$$
\boxed{\text{Overfitting}\approx\text{High Variance}}
$$

Overfitting and high variance are commonly used together to describe models that fit training data too closely and do not generalize well.

## 3. Underfitting vs Good Fit vs Overfitting:

To understand Overfitting, we should also understand Underfitting.

### A. Underfitting (High Bias):

Underfitting occurs when a model is too simple to capture the actual patterns in the training data.

For example, fitting a straight line to a dataset that follows a complicated curved relationship.

Characteristics:

- Model is too simple.
- Fails to capture important patterns.
- Has high training error.
- Has high validation error.
- Associated with High Bias.

Therefore:

$$
\boxed{\text{Underfitting}\approx\text{High Bias}}
$$

### B. Good Fit:

A good model learns the underlying patterns in the training data without memorizing unnecessary variations.

Characteristics:

- Appropriate model complexity.
- Relatively low training error.
- Low validation error.
- Performs well on unseen examples.
- Generalizes well.

### C. Overfitting (High Variance):

Overfitting occurs when a model is too complex and learns the training data too closely, including noise and unnecessary variations.

Characteristics:

- Model is too complex.
- Fits individual training examples too closely.
- Very low training error.
- High validation error.
- Performs poorly on unseen examples.
- Associated with High Variance.

### Comparison:

| Underfitting | Good Fit | Overfitting |
|---|---|---|
| Model too simple. | Appropriate complexity. | Model too complex. |
| High bias. | Balanced fit. | High variance. |
| High training error. | Low training error. | Very low training error. |
| High validation error. | Low validation error. | High validation error. |
| Does not learn enough. | Learns general patterns. | Memorizes training data too closely. |

**Important Relationships:**

$$
\boxed{\text{Underfitting}\approx\text{High Bias}}
$$

$$
\boxed{\text{Overfitting}\approx\text{High Variance}}
$$

These are commonly used associations rather than exact mathematical identities.

## 4. Why Does Overfitting Happen?

Overfitting is more likely when a model is too complex relative to the amount of available training data.

For example, suppose we are predicting house prices using Polynomial Regression.

Our original model is:

$$
f(x)=w_1x+b
$$

We then add more polynomial features:

$$
f(x)=w_1x+w_2x^2+w_3x^3+b
$$

We can also introduce higher-order features:

$$
x^4,\quad x^5,\quad x^6,\ldots
$$

As we add more polynomial features, the model becomes more flexible.

It has more parameters that can be adjusted to fit the training examples.

If we have very few training examples but use a highly complicated model, the model may start fitting individual variations instead of learning the general relationship.

This can produce a very wiggly curve and cause Overfitting.

### Example: Too Many Features:

Suppose we want to predict house prices using 100 different features.

Some examples include:

- House size.
- Number of bedrooms.
- Number of floors.
- Age of the house.
- Average income of the neighborhood.
- Distance to the nearest coffee shop.
- Distance to the nearest school.

If we have many features but insufficient training examples, our learning algorithm may overfit the training data.

**Main Idea:** A highly complicated model with insufficient training data is more likely to overfit.

# Solutions to Overfitting:

There are three main techniques for addressing Overfitting:

1. Collect more training data.
2. Use a subset of the most useful features (Feature Selection).
3. Apply Regularization.

## 5. Solution 1: Collect More Training Data:

One effective way to reduce Overfitting is to collect more training examples.

Suppose we initially have only a few house records containing:

- House size.
- House price.

A complicated polynomial model may fit these limited training examples too closely.

Now suppose we collect many more training examples.

The model receives more information about the actual relationship between house size and house price.

This can help it learn a smoother function that captures the general pattern rather than fitting every individual variation.

### Why Does More Data Help?

1. More examples provide additional information about the actual relationship.
2. The model becomes less dependent on individual training examples.
3. Unnecessary variations become less influential.
4. The model may generalize better to unseen examples.

Even a higher-order polynomial model may perform well when enough representative training examples are available.

### Important Point:

Collecting more training data is not always possible.

For example, there may be only a limited number of houses sold in a particular location.

When additional data is unavailable, we can use other techniques to reduce Overfitting.

**Conclusion:** When more representative training data is available, collecting it is a useful way to reduce Overfitting.

## 6. Solution 2: Feature Selection:

Feature Selection is the process of choosing the most useful features from a dataset and removing unnecessary or less useful features.

Suppose we have 100 features for predicting house prices.

Instead of using all 100 features, we select only the most relevant ones:

1. House size.
2. Number of bedrooms.
3. Age of the house.

Our model becomes:

$$
f_{\vec{w},b}(\vec{x})=
w_1x_1+w_2x_2+w_3x_3+b
$$

We have reduced the number of input features from 100 to 3.

### Why Does Feature Selection Help?

Using fewer features reduces the complexity of the model.

When there are too many features and insufficient training examples, the model may learn unnecessary relationships.

By selecting a smaller subset of relevant features, we can reduce the risk of Overfitting.

### How Do We Select Features?

One simple approach is to use domain knowledge and intuition.

For example, when predicting house prices:

- House size is likely to be relevant.
- Number of bedrooms is likely to be relevant.
- Age of the house is likely to be relevant.

We can select features that we believe are useful for the prediction task.

More advanced automatic feature-selection techniques will be covered later in the specialization.

### Disadvantage of Feature Selection:

Feature Selection has one important disadvantage.

By removing features, we also remove the information contained in those features.

For example, suppose all 100 features actually provide useful information about house prices.

If we keep only three features, we may lose some useful information and reduce the model's predictive performance.

Therefore, Feature Selection should be performed carefully.

**Conclusion:** Feature Selection reduces model complexity by using only a subset of the available features, but it may discard useful information.

## 7. Solution 3: Regularization:

Regularization is a technique used to reduce Overfitting by encouraging the learning algorithm to keep model parameter values relatively small.

It reduces the influence of parameters without necessarily setting them exactly to zero.

### Why Is Regularization Needed?

Consider a Polynomial Regression model:

$$
f(x)=w_1x+w_2x^2+w_3x^3+w_4x^4+b
$$

The model contains four weights:

$$
w_1,w_2,w_3,w_4
$$

When a model overfits, its parameters can sometimes become relatively large.

Large parameter values can make the model excessively sensitive to changes in input features, contributing to complicated and wiggly curves.

Regularization helps control this problem by encouraging smaller parameter values.

### Understanding Feature Removal:

Suppose we completely remove the feature $x^4$ from our model.

Originally:

$$
f(x)=w_1x+w_2x^2+w_3x^3+w_4x^4+b
$$

After removing $x^4$:

$$
f(x)=w_1x+w_2x^2+w_3x^3+b
$$

Mathematically, removing the contribution of $x^4$ is equivalent to:

$$
\boxed{w_4=0}
$$

This means that the fourth feature no longer contributes to the prediction.

This is similar to Feature Selection.

### How Is Regularization Different?

Instead of completely removing features, Regularization encourages the algorithm to shrink the parameter values.

For example, suppose the original weights are:

$$
\vec{w}=[100,250,400,500]
$$

After regularization, they might become:

$$
\vec{w}=[12,18,25,30]
$$

These are illustrative values.

Notice that all four features are still present.

Their weights are simply smaller.

### Main Idea of Regularization:

Regularization allows us to keep all the available features while reducing their excessive influence on the model's predictions.

It makes the model less likely to fit unnecessary variations in the training data.

### Which Parameters Are Regularized?

Our model contains:

$$
w_1,w_2,\ldots,w_n,b
$$

By convention, regularization is normally applied to:

$$
\boxed{w_1,w_2,\ldots,w_n}
$$

The bias parameter $b$ is usually not regularized.

In practice, regularizing or not regularizing the bias often makes very little difference in standard applications.

**Important:** The exact mathematical formula for Regularization will be covered in the next lecture.

## 8. Feature Selection vs Regularization:

Both techniques help reduce Overfitting, but they work differently.

| Feature Selection | Regularization |
|---|---|
| Removes less useful features. | Keeps the features but reduces the influence of their weights. |
| Reduces the number of features. | Encourages smaller parameter values. |
| May discard useful information. | Allows the model to retain all features. |
| Can set a feature's contribution to zero by removing it. | Does not necessarily set weights to exactly zero. |

### Example:

Suppose:

$$
\vec{w}=[100,250,400,500]
$$

**Feature Selection:**

Remove the fourth feature:

$$
\vec{w}=[100,250,400,0]
$$

**Regularization:**

Encourage all parameter values to become smaller:

$$
\vec{w}=[12,18,25,30]
$$

The second example keeps all features but controls their influence.

## 9. Comparison of the Three Solutions to Overfitting:

| Technique | What It Does | Main Advantage | Limitation |
|---|---|---|---|
| More Training Data | Adds more representative training examples. | Provides more information to the model. | More data may not be available. |
| Feature Selection | Uses only a subset of useful features. | Reduces model complexity. | May discard useful information. |
| Regularization | Encourages smaller model weights. | Keeps features while controlling their influence. | Requires choosing a suitable regularization strength. |

These methods can also be used together.

## 10. Important Concepts Summary:

### Overfitting:

$$
\boxed{\text{Overfitting}\approx\text{High Variance}}
$$

### Underfitting:

$$
\boxed{\text{Underfitting}\approx\text{High Bias}}
$$

### Three Ways to Reduce Overfitting:

$$
\boxed{
\begin{aligned}
&1.\ \text{Collect More Training Data}\\
&2.\ \text{Feature Selection}\\
&3.\ \text{Regularization}
\end{aligned}
}
$$

$$
\boxed{w_1,w_2,\ldots,w_n}
$$

The bias parameter $b$ is usually excluded from regularization.
