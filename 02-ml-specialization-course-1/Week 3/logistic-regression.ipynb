# **Logistic Regression:**

## 1. Introduction to Classification:

Classification is a supervised learning problem in which the objective is to predict a category or class rather than a numerical value.

For example, suppose we want to classify tumors into two categories:

- Malignant (Cancerous)
- Benign (Non-cancerous)

We assign numerical labels to these categories:

| Tumor Type | Label | Meaning |
|------------|-------|---------|
| Malignant | 1 | Positive class |
| Benign | 0 | Negative class |

Therefore:

$$
y\in\{0,1\}
$$

The output can only be 0 or 1.

**Important:** Positive class does not necessarily mean something good. It represents the class we want to detect.

### Regression vs Classification:

| Regression | Classification |
|------------|----------------|
| Predicts numerical values. | Predicts categories. |
| Output can be continuous. | Actual output belongs to a class. |
| Example: House price prediction. | Example: Tumor classification. |
| Linear Regression is commonly used. | Logistic Regression is commonly used. |

## 2. Why Can't We Use Linear Regression for Classification?

Recall the Linear Regression model:

$$
f_{w,b}(x)=wx+b
$$

Linear Regression can produce any real number, including negative values and values greater than 1.

However, in binary classification, the actual labels are:

$$
y=0\quad\text{or}\quad y=1
$$

Suppose a Linear Regression model produces:

- 0.2
- 0.7
- 1.3
- -0.4

The outputs 1.3 and -0.4 cannot be interpreted as probabilities because probabilities must be between 0 and 1.

Therefore, ordinary Linear Regression is not suitable for this classification problem.

We need a function that converts any real-valued input into an output between 0 and 1.

This is achieved using the **Sigmoid Function**.

## 3. Sigmoid Function:

The Sigmoid Function is a mathematical function that converts any real number into an output between 0 and 1.

It is also called the Logistic Function.

### Formula:

$$
\boxed{g(z)=\frac{1}{1+e^{-z}}}
$$

Where:

- $z$ = Input to the sigmoid function.
- $e$ = Mathematical constant approximately equal to 2.71828.
- $g(z)$ = Output of the sigmoid function.

The sigmoid function produces an S-shaped curve.

### Understanding the Behavior of the Sigmoid Function:

The sigmoid function behaves differently depending on the value of $z$.

**Case 1: When z = 0**

$$
g(0)=\frac{1}{1+e^{-0}}
$$

Since:

$$
e^0=1
$$

Therefore:

$$
g(0)=\frac{1}{1+1}
$$

$$
\boxed{g(0)=0.5}
$$

Whenever $z=0$, the sigmoid output is exactly 0.5.

**Case 2: When z is a large positive number**

Suppose:

$$
z=100
$$

Then:

$$
g(100)=\frac{1}{1+e^{-100}}
$$

Since $e^{-100}$ is extremely close to zero:

$$
g(100)\approx\frac{1}{1+0}
$$

$$
\boxed{g(100)\approx1}
$$

Therefore, when $z$ is a very large positive number, the sigmoid output approaches 1.

**Case 3: When z is a large negative number**

Suppose:

$$
z=-100
$$

Then:

$$
g(-100)=\frac{1}{1+e^{-(-100)}}
$$

$$
=\frac{1}{1+e^{100}}
$$

Since $e^{100}$ is extremely large:

$$
\boxed{g(-100)\approx0}
$$

Therefore, when $z$ is a very large negative number, the sigmoid output approaches 0.

### Summary

| Input $z$ | Sigmoid Output $g(z)$ |
|-----------|-----------------------|
| Very large negative | Close to 0 |
| 0 | 0.5 |
| Very large positive | Close to 1 |

The sigmoid function approaches 0 and 1 but never reaches either value for a finite input.



## 4. Building the Logistic Regression Model:

Andrew builds the Logistic Regression model in two steps.

### Step 1: Calculate the Linear Function:

First, we calculate:

$$
\boxed{z=\vec{w}\cdot\vec{x}+b}
$$

For one feature:

$$
z=wx+b
$$

For multiple features:

$$
z=w_1x_1+w_2x_2+\cdots+w_nx_n+b
$$

Where:

- $\vec{x}$ = Input feature vector.
- $\vec{w}$ = Weight vector.
- $b$ = Bias.
- $z$ = Linear calculation or score.

At this stage, $z$ can be any real number, including negative values and values greater than 1.

### Step 2: Apply the Sigmoid Function:

After calculating $z$, we pass it to the sigmoid function:

$$
g(z)=\frac{1}{1+e^{-z}}
$$

The sigmoid function converts the value of $z$ into an output between 0 and 1.

### Complete Logistic Regression Model:

Combining both steps:

$$
z=\vec{w}\cdot\vec{x}+b
$$

And:

$$
g(z)=\frac{1}{1+e^{-z}}
$$

Therefore:

$$
\boxed{
f_{\vec{w},b}(\vec{x})=
g(\vec{w}\cdot\vec{x}+b)
}
$$

Expanding the sigmoid function:

$$
\boxed{
f_{\vec{w},b}(\vec{x})=
\frac{1}{1+e^{-(\vec{w}\cdot\vec{x}+b)}}
}
$$

This is the Logistic Regression model.

**How it works:**

1. Take the input features.
2. Multiply each feature by its corresponding weight.
3. Add the bias.
4. Calculate $z$.
5. Pass $z$ into the sigmoid function.
6. Obtain an output between 0 and 1.

## 5. Numerical Example of Logistic Regression:

Suppose we want to predict whether a tumor is malignant based on its size.

Assume:

$$
w=0.5
$$

$$
b=-2
$$

And the tumor size is:

$$
x=6
$$

### Step 1: Calculate z

Using:

$$
z=wx+b
$$

Substitute:

$$
z=0.5(6)-2
$$

$$
=3-2
$$

$$
\boxed{z=1}
$$

### Step 2: Apply the Sigmoid Function

Using:

$$
g(z)=\frac{1}{1+e^{-z}}
$$

Substitute $z=1$:

$$
g(1)=\frac{1}{1+e^{-1}}
$$

We know:

$$
e^{-1}\approx0.3679
$$

Therefore:

$$
g(1)=\frac{1}{1+0.3679}
$$

$$
=\frac{1}{1.3679}
$$

$$
\boxed{g(1)\approx0.731}
$$

Therefore, the Logistic Regression model produces an output of approximately 0.731.

This is interpreted as an estimated 73.1% probability that the tumor belongs to the malignant class.

## 6. Interpretation of Logistic Regression Output:

The output of Logistic Regression is interpreted as the probability that the class label $y$ is equal to 1, given the input features.

Mathematically:

$$
\boxed{
f_{\vec{w},b}(\vec{x})=
P(y=1\mid\vec{x};\vec{w},b)
}
$$

For example, suppose the model outputs:

$$
f_{\vec{w},b}(\vec{x})=0.7
$$

This means:

$$
P(y=1\mid\vec{x})=0.7
$$

In simple words, the model estimates a 70% probability that the tumor is malignant.

**Important Distinction:**

- The model output can be a probability such as 0.7.
- The actual class label can only be 0 or 1.

Therefore, 0.7 is not the actual class label. It is the predicted probability of belonging to class 1.

| Concept | Value | Meaning |
|---------|-------|---------|
| Model output | 0.7 | Estimated probability |
| Actual label | 0 | Benign |
| Actual label | 1 | Malignant |

Logistic Regression produces probabilities, which can later be converted into class predictions using a decision rule.

## 7. Relationship Between the Probabilities of Class 0 and Class 1

In binary classification, there are only two possible classes:

$$
y\in\{0,1\}
$$

The probabilities of these two classes must add up to 1.

Therefore:

$$
P(y=0\mid x)+P(y=1\mid x)=1
$$

Rearranging:

$$
\boxed{
P(y=0\mid x)=1-P(y=1\mid x)
}
$$

### Numerical Example

Suppose:

$$
P(y=1\mid x)=0.7
$$

Then:

$$
P(y=0\mid x)=1-0.7
$$

$$
\boxed{P(y=0\mid x)=0.3}
$$

Therefore:

- Probability of malignant = 70%.
- Probability of benign = 30%.

The total probability is:

$$
0.7+0.3=1
$$

$$
=100\%
$$

## 8. Understanding Probability Notation

In research papers and machine learning articles, you may see the following notation:

$$
\boxed{
f_{\vec{w},b}(\vec{x})=
P(y=1\mid\vec{x};\vec{w},b)
}
$$

Meaning of each symbol:

| Symbol | Meaning |
|--------|---------|
| $P$ | Probability |
| $y=1$ | Output belongs to the positive class |
| $\mid$ | Given |
| $\vec{x}$ | Input features |
| $\vec{w}$ | Model weights |
| $b$ | Model bias |
| Semicolon | Separates the model parameters in this notation |

Read the expression as:

"The probability that the output is 1, given the input features and the model parameters."

This notation represents the probability estimated by the Logistic Regression model.

## 9. Decision Threshold:

Logistic Regression produces probabilities between 0 and 1.

However, we often need a final class prediction of either 0 or 1.

A Decision Threshold is used to convert predicted probabilities into class predictions.

Andrew uses a common threshold:

$$
\boxed{\text{Threshold}=0.5}
$$

The decision rule is:

$$
\boxed{
\begin{cases}
\hat{y}=1 & \text{if } f_{\vec{w},b}(\vec{x})\geq0.5\\
\hat{y}=0 & \text{if } f_{\vec{w},b}(\vec{x})<0.5
\end{cases}
}
$$

Where $\hat{y}$ represents the predicted class.

### Examples

| Predicted Probability | Predicted Class |
|-----------------------|-----------------|
| 0.9 | 1 |
| 0.7 | 1 |
| 0.5 | 1 |
| 0.4 | 0 |
| 0.2 | 0 |

**Important:** Exactly 0.5 is classified as class 1 because the rule uses greater than or equal to 0.5.

## 10. Connection Between the Decision Threshold and Sigmoid Function:

Recall:

$$
f(x)=g(z)
$$

The decision rule says:

$$
\hat{y}=1\quad\text{when}\quad f(x)\geq0.5
$$

Replacing $f(x)$ with $g(z)$:

$$
g(z)\geq0.5
$$

From the sigmoid function, we know:

$$
g(z)=\frac{1}{1+e^{-z}}
$$

The sigmoid output is greater than or equal to 0.5 whenever:

$$
\boxed{z\geq0}
$$

Similarly, the sigmoid output is less than 0.5 whenever:

$$
\boxed{z<0}
$$

Since:

$$
z=\vec{w}\cdot\vec{x}+b
$$

The final decision rules become:

**Predict class 1:**

$$
\boxed{\vec{w}\cdot\vec{x}+b\geq0}
$$

**Predict class 0:**

$$
\boxed{\vec{w}\cdot\vec{x}+b<0}
$$

This allows us to determine the predicted class by checking whether $z$ is positive or negative.

## 11. Decision Boundary:

A Decision Boundary is a line or curve that separates the regions where a Logistic Regression model predicts class 0 and class 1.

The decision boundary occurs when:

$$
\boxed{z=0}
$$

Since:

$$
z=\vec{w}\cdot\vec{x}+b
$$

The decision boundary equation is:

$$
\boxed{\vec{w}\cdot\vec{x}+b=0}
$$

At the decision boundary:

$$
g(0)=0.5
$$

Therefore, the model is exactly at the default decision threshold.

One side of the boundary predicts class 0, while the other side predicts class 1.


## 12. Decision Boundary with Two Features:

Suppose we have two input features:

- $x_1$ = First feature.
- $x_2$ = Second feature.

The Logistic Regression model is:

$$
z=w_1x_1+w_2x_2+b
$$

Andrew uses these parameters:

$$
w_1=1,\quad w_2=1,\quad b=-3
$$

Substitute:

$$
z=1x_1+1x_2-3
$$

Simplify:

$$
z=x_1+x_2-3
$$

### Calculate the Decision Boundary

We know that the decision boundary occurs when:

$$
z=0
$$

Therefore:

$$
x_1+x_2-3=0
$$

Rearranging:

$$
\boxed{x_1+x_2=3}
$$

This is the equation of a straight line.

### Points on the Decision Boundary

| $x_1$ | $x_2$ |
|------:|------:|
| 0 | 3 |
| 1 | 2 |
| 2 | 1 |
| 3 | 0 |

All these points satisfy:

$$
x_1+x_2=3
$$

### Predictions on Different Sides

**Case A: Positive class**

Given:

$$
x_1=2,\quad x_2=2
$$

Calculate:

$$
z=2+2-3
$$

$$
z=1
$$

Since:

$$
z\geq0
$$

Prediction:

$$
\boxed{\hat{y}=1}
$$

**Case B: Negative class**

Given:

$$
x_1=1,\quad x_2=1
$$

Calculate:

$$
z=1+1-3
$$

$$
z=-1
$$

Since:

$$
z<0
$$

Prediction:

$$
\boxed{\hat{y}=0}
$$

**Case C: Exactly on the decision boundary**

Given:

$$
x_1=1,\quad x_2=2
$$

Calculate:

$$
z=1+2-3=0
$$

Therefore:

$$
g(0)=0.5
$$

Since the threshold is 0.5:

$$
\boxed{\hat{y}=1}
$$



## 13. Nonlinear Decision Boundaries Using Polynomial Features:

A decision boundary does not always have to be a straight line.

We can introduce polynomial features into Logistic Regression, just like we did in Polynomial Regression.

This allows us to create curved and more complex decision boundaries.

### Example: Circular Decision Boundary:

Andrew chooses the following model:

$$
z=w_1x_1^2+w_2x_2^2+b
$$

The parameters are:

$$
w_1=1,\quad w_2=1,\quad b=-1
$$

Substitute:

$$
z=1x_1^2+1x_2^2-1
$$

Simplify:

$$
\boxed{z=x_1^2+x_2^2-1}
$$

### Calculate the Decision Boundary:

We know:

$$
z=0
$$

Therefore:

$$
x_1^2+x_2^2-1=0
$$

Rearranging:

$$
\boxed{x_1^2+x_2^2=1}
$$

This is the equation of a circle centered at the origin with radius 1.

### Predictions Inside and Outside the Circle:

**Inside the circle:**

$$
x_1^2+x_2^2<1
$$

Therefore:

$$
z<0
$$

Prediction:

$$
\boxed{\hat{y}=0}
$$

**Outside the circle:**

$$
x_1^2+x_2^2>1
$$

Therefore:

$$
z>0
$$

Prediction:

$$
\boxed{\hat{y}=1}
$$

**Exactly on the circle:**

$$
x_1^2+x_2^2=1
$$

Therefore:

$$
z=0
$$

Prediction:

$$
\boxed{\hat{y}=1}
$$

### Numerical Examples

Point A:

$$
(x_1,x_2)=(0,0)
$$

$$
z=0^2+0^2-1=-1
$$

Prediction:

$$
\boxed{\hat{y}=0}
$$

Point B:

$$
(x_1,x_2)=(1,0)
$$

$$
z=1^2+0^2-1=0
$$

Prediction:

$$
\boxed{\hat{y}=1}
$$

Point C:

$$
(x_1,x_2)=(2,0)
$$

$$
z=2^2+0^2-1=3
$$

Prediction:

$$
\boxed{\hat{y}=1}
$$


## 14. Linear vs Nonlinear Decision Boundaries:

The shape of the decision boundary depends on the features used by the model.

| Features Used | Possible Decision Boundary |
|---------------|----------------------------|
| $x_1,x_2$ | Straight line |
| $x_1^2,x_2^2$ | Circle or ellipse, depending on parameters |
| $x_1,x_2,x_1^2,x_1x_2,x_2^2$ | General quadratic curve |
| Higher-order polynomial features | More complex curves |
